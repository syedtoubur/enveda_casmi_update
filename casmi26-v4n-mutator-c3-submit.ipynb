# CASMI26 | v4n + Engine Fusion + Union — public LB 0.399

The same recipe that took v4g from 0.366 to 0.386, applied on top of the stronger public **v4n** base:

| Step | Public LB |
|---|---|
| v4n alone (@ahmedberatozer: v4g + GLACIER re-scoring + PubChem N1 5000 + ICE/GL weight 1.0 + fpnet_full1 engine bank) | 0.384 |
| + our two-ranker engine (BIO + AFIX) as a second list, weighted reciprocal-rank fusion α = 0.6, K = 3 | — |
| + **forward models on the union**: the engine's candidates are also scored by ICEBERG and GLACIER, and the fused list is re-ranked inside same-formula groups by `z(RRF) + 1.0·z(ICEBERG) + 1.0·z(GLACIER)` | **0.399** |

For reference, on the older v4g base the same steps gave 0.366 → 0.382 (fusion) → 0.386 (union).

Why it works: the two retrieval systems use different candidate pools and rankers, so their top-25 lists make different
mistakes and a rank fusion beats either one. The fusion however pulls engine candidates into the final list that the base never
sent to the forward models, so their isomer order would come from retrieval scores alone; scoring the union fixes that.

Things that did **not** help (public LB, v4g-based variant at 0.386): forward-model weight 2.0 everywhere → 0.352 (too much
forward weight destroys confident library matches); same-formula window TOPN 100 instead of 60 → 0.386 (no change).

**Runtime:** ~3.5–4.5 h on T4 for the hidden test (budgets: ICEBERG 5400 s, GLACIER 4000 s), internet off.

**Commit vs. scoring rerun.** The visible `test.parquet` is a dummy, so a commit only runs the first 12 molecules
(fast, produces a valid `submission.csv`); the competition rerun on the hidden test is detected automatically
(`KAGGLE_IS_COMPETITION_RERUN` or a different test signature) and runs everything. Fork → *Save Version* → *Submit*.

## Our engine first (two-ranker + BIO + AFIX)
Runs in its own process before v4g (GPU memory is still empty) and writes its top-40 per molecule with tautomer-canonical

- **BIO**: ChEBI + LIPID MAPS structures added to the candidate pool (public LB 0.341 → 0.350 alone).

## Mutator channel — Class-3 routed (specs/track1_retrieval_fusion §10)
Bemis-Murcko scaffolds from each molecule's PubChem top-25, its engine-A top-40 (BIO = ChEBI + LIPID MAPS, AFIX analog propagation) **and its mass-shifted spectral analogs**, decorated to a target formula by the 20-transform biogenetic palette in formula space, screened by the same two-pass `f.z` the PubChem channel uses.

Two things differ from `casmi26-v4n-mutator-bio-submit.ipynb` (LB 0.397):

1. **It fires on a Class-3 judgement, not on a fixed 30% of the field.** `c3_gate` wants a strong mass-shifted analog, a weak exact-mass match, and a formula delta within `C3_MAX_STEPS` transforms of that analog. All three are similarities in [0,1] or counts, so none depends on the `f.z` scale — which matters because a commit run scores a slice of `train.parquet` and cannot calibrate an absolute `f.z` cut. Firing on zero molecules is a legal outcome.
2. **Promotion replaces fusion.** Generated structures are same-formula regioisomers of the target, so in the bio notebook they always landed in the correct answer's own formula group, where `z(RRF)+z(ICE)+z(GL)` could swap the two — one such event costs 0.95 RR, the whole of 0.399 → 0.397. Here the mutator is absent from the fusion and is spliced into ranks reserved by tier instead, and only where the gate fired. On a true Class-3 molecule the retrieval list holds no truth, so those ranks cost nothing.

Measured on `p1_novel/pseudo_c3.parquet` (600 leak-free Class-3 natural products), restricted to scaffolds that are actually harvestable: depth ≤ 1 fires on 2.5% with a 80.0% chance the truth is generated, against 11.8% and 28.2% at depth ≤ 2 — the same expected gain from four times the exposure. Any failure leaves the 0.399 pipeline untouched.

v4n top-N + engine top-40). v4n's GLACIER runs on the same input, so the engine candidates get both forward scores.
v4n's own re-ranking and gated PubChem merge then write the v4n `submission.csv`.

```python
# ICEBERG post-ranker isomer re-scoring (casmi26-iceberg). Optional: any failure keeps the ranker order.
import gc
MUT = globals().get('MUT') or {}
ICE_SCORES, ice_stats = {}, dict(molecules=0, changed_top25=0, changed_top1=0)
try:
    ICE_PKG = os.path.dirname(find('casmi26-iceberg/**/ice_runner.py'))
    sys.path.append(ICE_PKG)
    import fuse as ice_fuse
    del V, E, bank; gc.collect(); torch.cuda.empty_cache()
    # priority: molecules without a library-level match first (class 1 rarely changes), then the rest
    mids = sorted(BASE, key=lambda m: BASE[m][2])
    cands = {m: ice_fuse.ice_candidates(BASE[m][0], BASE[m][4], TOPN) for m in mids}
    if ENG:                              # ADDED (ICE_UNION): also score our engine's candidates (same-formula groups
        from rdkit.Chem.rdMolDescriptors import CalcMolFormula as _CMF   # formed over the v4g top-N + engine top-40)
        from rdkit import Chem as _Ch
        def _f2(s):
            try:
                return _CMF(_Ch.MolFromSmiles(s))
            except Exception:
                return s
        n_add = 0
        for m in mids:
            e = ENG.get(str(m))
            if not e or not e.get('smiles'):
                continue
            bs, bk, bf = list(BASE[m][0][:TOPN]), set(BASE[m][1][:TOPN]), list(BASE[m][4][:TOPN])
            es = [s for s, k in zip(e['smiles'][:40], e['keys'][:40]) if k not in bk]
            ef = [_f2(s) for s in es]
            extra = ice_fuse.ice_candidates(bs + es, bf + ef, len(bs) + len(es))
            before = len(cands.get(m, []))
            cands[m] = list(dict.fromkeys(list(cands.get(m, [])) + extra))
            n_add += len(cands[m]) - before
        print('ICE_UNION: engine candidates added to ICE input:', n_add, flush=True)
    if MUT:                              # ADDED (ICE_UNION_MUT): the mutator's candidates too.
        n_mut_add = 0                    # mandatory, not an optimisation -- generated structures have
        for m in mids:                   # no retrieval score, so ICE/GL are the only signal that can
            u = MUT.get(str(m))          # order them, and they are same-formula regioisomers.
            if not u or not u.get('smiles'):
                continue
            _known = set(BASE[m][1][:TOPN]) | set((ENG.get(str(m)) or {}).get('keys', []))
            _ms = [s for s, k in zip(u['smiles'][:25], u['keys'][:25]) if k not in _known]
            _mf = [f for f, k in zip(u['formula'][:25], u['keys'][:25]) if k not in _known]
            if not _ms:
                continue
            _before = len(cands.get(m, []))
            cands[m] = list(dict.fromkeys(list(cands.get(m, [])) + ice_fuse.ice_candidates(_ms, _mf, len(_ms))))
            n_mut_add += len(cands[m]) - _before
        print('ICE_UNION_MUT: mutator candidates added to ICE input:', n_mut_add, flush=True)
    if ICE_PC:                          # also the PubChem-only list of molecules the v3.6 gate may use (lib_max < LIB_TAU)
        from rdkit.Chem.rdMolDescriptors import CalcMolFormula
        from rdkit import Chem as _Chem
        def _form(s):
            try:
                return CalcMolFormula(_Chem.MolFromSmiles(s))
            except Exception:
                return s
        for m in mids:
            p = PC.get(m) if isinstance(PC, dict) else None
            if p and p.get('pc') and BASE[m][2] < LIB_TAU:
                p['pc_form'] = [_form(x) for x in p['pc']]
                extra = ice_fuse.ice_candidates(p['pc'], p['pc_form'], 25)
                cands[m] = list(dict.fromkeys(list(cands.get(m, [])) + extra))
    cands = {m: c for m, c in cands.items() if c}
    items = ice_fuse.build_ice_input(te, cands)
    print('ICE input molecules', len(items), 'candidates', sum(len(it['cands']) for it in items), f'{time.time()-T0:.0f}s', flush=True)
    ICE_SCORES = ice_fuse.run_ice(ICE_PKG, items, workdir='/kaggle/working/ice_work', device='cuda', budget_s=ICE_BUDGET,
                                  site='/kaggle/working/ice_site')
    meta_f = '/kaggle/working/ice_work/ice_out.json.meta.json'
    if os.path.exists(meta_f):
        print('ICE meta', open(meta_f).read()[:1500])
except Exception as e:
    print('ICE FAILED -> ranker order kept:', repr(e))
# v4h: GLACIER third re-scoring term (casmi26-glacier) on the SAME runner input; any GLACIER failure -> v4g (ICE only)
GL_SCORES, gl_stats, GL_LAM, GL_BUDGET = {}, dict(molecules=0, changed_vs_ice_top25=0, changed_vs_ice_top1=0), 1.0, 5200
try:
    GL_PKG = os.path.dirname(find('casmi26-glacier/**/gl_runner.py'))
    sys.path.append(GL_PKG)
    import gl_fuse
    gc.collect(); torch.cuda.empty_cache()   # ICE ran in its own (finished) process; free the notebook's cached GPU memory
    GL_SCORES = gl_fuse.run_gl(GL_PKG, items, workdir='/kaggle/working/gl_work', device='cuda',
                               budget_s=(GL_BUDGET_SMOKE if not IS_RERUN else GL_BUDGET),
                               site='/kaggle/working/ice_site', wheels=os.path.join(ICE_PKG, 'wheels'))
    meta_f = '/kaggle/working/gl_work/gl_out.json.meta.json'
    print('GL meta', open(meta_f).read()[:1500] if os.path.exists(meta_f) else 'missing')
except Exception as e:
    GL_SCORES = {}
    print('GL FAILED -> v4g behaviour (ICE only):', repr(e))
def gl_of(m):
    g = GL_SCORES.get(str(m), {}) if isinstance(GL_SCORES, dict) else {}
    return g if isinstance(g, dict) and any(v is not None for v in g.values()) else {}
def gl_rerank(order0, smis, keys, scs, forms, ice, gl, top_n, tag=''):
    # z(scs) + ICE_LAM z(ice) + GL_LAM z(gl) in the same slots; no GL scores / any failure -> order0 (= v4g order)
    if not gl:
        return order0
    try:
        o = gl_fuse.rerank_multi(smis, keys, scs, forms, [ice, gl], [ICE_LAM, GL_LAM], top_n=top_n)
    except Exception as e:
        print('gl rerank failed', tag, repr(e)); return order0
    for k, v in (('molecules', 1), ('changed_vs_ice_top25', int(o[:25] != order0[:25])), ('changed_vs_ice_top1', int(o[:1] != order0[:1]))):
        gl_stats[tag + k] = gl_stats.get(tag + k, 0) + v
    return o
for m in list(BASE):
    smis, keys, lib_max, scs, forms = BASE[m]
    ice = ICE_SCORES.get(str(m), {}) if ICE_SCORES else {}
    gl = gl_of(m)
    if ice and any(v is not None for v in ice.values()):
        try:
            order = ice_fuse.rerank(smis, keys, scs, forms, ice, lam=ICE_LAM, top_n=TOPN)
            order = gl_rerank(order, smis, keys, scs, forms, ice, gl, TOPN)
            s2 = [smis[i] for i in order]; k2 = [keys[i] for i in order]
            ice_stats['molecules'] += 1
            ice_stats['changed_top25'] += int(s2[:25] != smis[:25]); ice_stats['changed_top1'] += int(bool(s2) and s2[0] != smis[0])
            smis, keys = s2, k2
        except Exception as e:
            print('rerank failed', m, repr(e))
    BASE[m] = (smis[:25], keys[:25], lib_max)
    p = PC.get(m) if (ICE_PC and isinstance(PC, dict)) else None
    if p and p.get('pc_form') and ice:
        try:                             # re-order the PubChem list by z(f.z) + lam z(ice) inside same-formula groups;
            o = ice_fuse.rerank(p['pc'], p['pc_keys'], p['pc_fz'], p['pc_form'], ice, lam=ICE_LAM, top_n=25)
            o = gl_rerank(o, p['pc'], p['pc_keys'], p['pc_fz'], p['pc_form'], ice, gl, 25, 'pc_')
            fz0 = p['pc_fz'][0]          # the gate keeps using the ORIGINAL top f.z (rel), so gate decisions are unchanged
            p['pc'] = [p['pc'][i] for i in o]; p['pc_keys'] = [p['pc_keys'][i] for i in o]
            p['pc_fz'] = [fz0] + [p['pc_fz'][i] for i in o][1:]
            ice_stats['pc_changed'] = ice_stats.get('pc_changed', 0) + int(o[:1] != [0])
        except Exception as e:
            print('pc rerank failed', m, repr(e))
print('ICE rerank stats', ice_stats, f'{time.time()-T0:.0f}s')
print('GL rerank stats', gl_stats, f'{time.time()-T0:.0f}s')
```

```python
# gated PubChem merge (v3.6, unchanged) -> submission.csv
def merge(base, base_keys, pc, pc_keys, slots, n=25):
    bk = set(base_keys)
    pcs = [s for s, k in zip(pc, pc_keys) if k not in bk]
    out, bi, pj = [], 0, 0
    for pos in range(1, n + 1):
        if pos in slots and pj < len(pcs):
            out.append(pcs[pj]); pj += 1
        elif bi < len(base):
            out.append(base[bi]); bi += 1
        elif pj < len(pcs):
            out.append(pcs[pj]); pj += 1
    return out
rows, stats = [], dict(untouched=0, gentle=0, aggressive=0, no_pc=0)
for mid in te.molecule_id.unique():
    smis, keys, lib_max = BASE.get(mid, ([], [], 0.0))
    p = PC.get(mid)
    if p is None or not p['pc']:
        stats['no_pc'] += 1; final = smis
    elif lib_max >= LIB_TAU:
        stats['untouched'] += 1; final = smis
    else:
        rel = p['pc_fz'][0] - p['best_pool_fz'] if np.isfinite(p['best_pool_fz']) else 1e9
        slots = SLOTS_AGG if rel > REL_TH else SLOTS_GENTLE
        stats['aggressive' if rel > REL_TH else 'gentle'] += 1
        final = merge(smis, keys, p['pc'], p['pc_keys'], slots)
    if not final: final = ['CCO']
    rows.append((mid, ';'.join(final[:25])))
print('merge stats', stats)
sub = pd.DataFrame(rows, columns=['molecule_id', 'smiles'])
samp = pd.read_csv(os.path.join(COMP, 'sample_submission.csv'))
sub = samp[['molecule_id']].merge(sub, on='molecule_id', how='left')
sub['smiles'] = sub['smiles'].fillna('CCO')
assert len(sub) == len(samp) and sub.molecule_id.is_unique and sub.smiles.notna().all()
assert sub.smiles.map(lambda s: len(s.split(';'))).max() <= 25
sub.to_csv('submission.csv', index=False)
json.dump(dict(secs=round(time.time() - T0), merge=stats, n_errors=n_err, families=MAN['families'],
               gpu=torch.cuda.get_device_name(0) if dev == 'cuda' else None), open('run_manifest.json', 'w'), indent=1)
print('done', sub.shape, f'{time.time()-T0:.0f}s')
sub.head()
```

## Fusion + forward re-rank
Weighted RRF of the v4n list (weight 1) and the engine list (α = 0.6), K = 3, top 40 → re-rank inside same-formula groups by
`z(RRF) + ICE_LAM·z(ICEBERG) + GL_LAM·z(GLACIER)` (both 1.0, as in v4n) → top 25.

```python

ALPHA, KRR, ALPHA_MUT = 0.6, 3.0, 0.25   # mutator below engine A: least trustworthy source
from rdkit.Chem.rdMolDescriptors import CalcMolFormula as _CMF2
from rdkit import Chem as _Ch2
def _form2(s):
    try:
        return _CMF2(_Ch2.MolFromSmiles(s))
    except Exception:
        return s
def fuse2(v_smis, e_smis, e_keys, m_smis=(), m_keys=(), n=40):
    sc, smi_of = {}, {}
    for r, s in enumerate(v_smis, 1):
        k = chem.score_key(s) or s
        sc[k] = sc.get(k, 0.0) + 1.0 / (KRR + r); smi_of.setdefault(k, s)
    for r, (s, k) in enumerate(zip(e_smis, e_keys), 1):
        if not k:
            continue
        sc[k] = sc.get(k, 0.0) + ALPHA / (KRR + r); smi_of.setdefault(k, s)
    for r, (s, k) in enumerate(zip(m_smis, m_keys), 1):      # ADDED: mutator list, weight ALPHA_MUT
        if not k:
            continue
        sc[k] = sc.get(k, 0.0) + ALPHA_MUT / (KRR + r); smi_of.setdefault(k, s)
    order = sorted(sc, key=lambda k: -sc[k])[:n]
    return [smi_of[k] for k in order], order, [sc[k] for k in order]

if ENG or MUT:
    v4 = pd.read_csv('submission.csv')
    changed = n_ice = 0; out = []
    mut_keys_all = {m: set(v['keys']) for m, v in MUT.items()}
    n_mut_in_top25 = n_mut_mols = 0
    C3_SLOTS = {'sure': [1, 2, 3], 'likely': [2, 3, 4]}
    n_c3 = n_c3_ice = 0; n_c3_tier = {}
    for mid, s in zip(v4.molecule_id, v4.smiles):
        vs = [x for x in s.split(';') if x and x != 'CCO']
        e = ENG.get(str(mid)) or {'smiles': [], 'keys': []}
        u = MUT.get(str(mid)) or {'smiles': [], 'keys': []}
        _tier = (u.get('tier') or '') if u.get('smiles') else ''
        if e['smiles']:
            fsm, fk, fsc = fuse2(vs, e['smiles'], e['keys'])
            ice = ICE_SCORES.get(str(mid), {}) if isinstance(ICE_SCORES, dict) else {}
            if ice:
                try:
                    o = ice_fuse.rerank(fsm, fk, fsc, [_form2(x) for x in fsm], ice, lam=ICE_LAM, top_n=len(fsm))
                    _g = gl_of(mid)
                    if _g:
                        o = gl_fuse.rerank_multi(fsm, fk, fsc, [_form2(x) for x in fsm], [ice, _g], [ICE_LAM, GL_LAM], top_n=len(fsm))
                    n_ice += int(o[:25] != list(range(min(25, len(o)))))
                    fsm = [fsm[i] for i in o]
                except Exception as ex:
                    print('fused ICE rerank failed', mid, repr(ex))
            changed += int(bool(vs) and fsm[:1] != vs[:1])
            vs = fsm[:25]
        if _tier:
            _ms, _mk = list(u['smiles']), list(u['keys'])
            _ice = ICE_SCORES.get(str(mid), {}) if isinstance(ICE_SCORES, dict) else {}
            _gl = gl_of(mid)
            if _ice:          # order the mutator's own candidates by forward-sim agreement. This is
                _sc = []      # legitimate where the fusion above was not: it ranks same-formula
                for _a, _b in zip(_ms, _mk):   # isomers against each other, never against a
                    _v = _ice.get(_b, _ice.get(_a, 0.0)) or 0.0      # retrieved candidate (R4).
                    if _gl:
                        _v += _gl.get(_b, _gl.get(_a, 0.0)) or 0.0
                    _sc.append(_v)
                if any(_sc):  # all-zero means neither key space matched, so keep the f.z order
                    _o = sorted(range(len(_ms)), key=lambda i: -_sc[i])
                    _ms = [_ms[i] for i in _o]; _mk = [_mk[i] for i in _o]
                    n_c3_ice += 1
            vs = merge(vs, [chem.score_key(x) or x for x in vs], _ms, _mk,
                       C3_SLOTS.get(_tier) or C3_SLOTS['likely'])
            n_c3 += 1; n_c3_tier[_tier] = n_c3_tier.get(_tier, 0) + 1
            _hit = sum(1 for x in vs[:25] if (chem.score_key(x) or x) in set(_mk))
            n_mut_in_top25 += _hit; n_mut_mols += int(_hit > 0)
        out.append((mid, ';'.join(vs[:25]) if vs else 'CCO'))
    fs = pd.DataFrame(out, columns=['molecule_id', 'smiles'])
    assert len(fs) == len(v4) and fs.molecule_id.is_unique and fs.smiles.notna().all()
    assert fs.smiles.map(lambda s: len(s.split(';'))).max() <= 25
    fs.to_csv('submission.csv', index=False)
    print('fused (+ICE) submission written; top-1 changed in', changed, '| ICE reordered', n_ice, f'{time.time()-T0:.0f}s')
    print(f'MUTATOR IN FINAL TOP-25: {n_mut_in_top25} candidates across {n_mut_mols} molecules')
    print(f'C3 ROUTER: promoted {n_c3} molecules {n_c3_tier} | forward-sim reordered the mutator list on {n_c3_ice} | suppressed on {len(MUT) - n_c3}')
```
