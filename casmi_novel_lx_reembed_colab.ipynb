# specs_novel — G0.1-R2: re-embed the corpus with the fixed `dreams_peaks` (Colab, GPU)

**Generated by `build/mk_novel_p0.py` — do not hand-edit.** Pre-registered in
`specs_novel/gate_report.md` → *G0.1-R2*. Precondition for P3.

G0.1-R failed because `np.argsort` let the platform choose which intensity-tied peaks survive the
top-100 cut. `dreams_peaks` now uses `np.lexsort` (ties go to the higher m/z). This notebook:

1. rebuilds every corpus row's peaks with the fixed code and flags rows with a **tie at the cut**;
2. picks **`parity_lx`** — 200 rows, 150 of them tie rows spread over libraries — and saves their
   peaks to `_novel/lx/parity_lx.npz` **before** embedding, so the Colab CPU leg
   (`casmi_novel_lx_cpu_colab.ipynb`) can start in parallel;
3. embeds all 1,323,878 rows into the **new** folder `_novel/emb_dbec3a0_lx/` (resumable; the old
   caches are only read);
4. records new vs `emb_dbec3a0` (the accepted risk, measured) and packages the Kaggle dataset
   **casmi26-novel-lx** for the Kaggle leg (`casmi_novel_lx_kaggle.ipynb`).

G0.1-R2 passes when this notebook's checks pass **and** the CPU and Kaggle legs rebuild
byte-identical peaks, **and** Kaggle's vectors reach cosine ≥ 0.999 with the new cache on every row.

Runtime: **GPU with High-RAM** (A100 ≈ 35 min; T4 ≈ 2.5 h).

This measures the risk accepted when G0 was closed by decision: how far the provisional cache is
from the fixed code. Tie rows are split out.

```python
t0 = time.time()
PRV = [np.load(PREV / f"emb_{i:04d}.npy", mmap_mode="r") for i in range(len(BOUNDS))]
COS = np.empty(N, np.float64)
for i, (lo, hi) in enumerate(BOUNDS):
    for a in range(0, hi - lo, 50_000):
        b = min(a + 50_000, hi - lo)
        COS[lo + a:lo + b] = lx.rowcos(np.asarray(NEW[i][a:b]), np.asarray(PRV[i][a:b]))
np.save(OUT / "cos_lx_vs_dbec3a0.npy", COS.astype(np.float32))

QS = [0.0, 0.001, 0.01, 0.1, 0.5]

def summarise(cos, groups):
    """Per group: n, min, median and the share below 0.999, worst groups first."""
    df = pd.DataFrame({"g": groups, "cos": cos})
    out = df.groupby("g").cos.agg(n="size", min="min", median="median",
                                  below_0999=lambda s: float((s < 0.999).mean()))
    return out.sort_values("min").reset_index()

overall = {f"q{q:g}": float(np.quantile(COS, q)) for q in QS}
overall.update(below_0999=float((COS < 0.999).mean()), below_0995=float((COS < 0.995).mean()),
               below_099=float((COS < 0.99).mean()), non_tie_min=float(COS[~TIE].min()) if (~TIE).any() else None)
BY = {"tie_at_cut": TIE.astype(int), "ingest_lib": LIB, "adduct": META.adduct.astype(str).to_numpy(),
      "oob_mass": META.oob_mass.to_numpy(), "npk_bin": np.digitize(NPK, [10, 40])}
by = {k: summarise(COS, v).to_dict("records") for k, v in BY.items()}
RESULT["lx_vs_dbec3a0"] = dict(overall=overall, by=by, seconds=round(time.time() - t0))
save_result()
print(json.dumps(overall, indent=1))
for k in BY:
    print(f"\nby {k}:\n{summarise(COS, BY[k]).head(12).to_string(index=False)}")
```
