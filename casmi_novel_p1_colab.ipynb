## Kernel packages: RDKit 2026.03.3 and the pinned selfies

```python
def kernel_version(mod):
    r = subprocess.run([sys.executable, "-c", f"import {mod}; print({mod}.__version__)"],
                       capture_output=True, text=True)
    return r.stdout.strip() if r.returncode == 0 else None

def v_match(v, target):
    if v is None:
        return False
    return v == target or v.replace(".0", ".") == target.replace(".0", ".")

if not v_match(kernel_version("rdkit"), RDKIT_PIN):
    py_tag = f"cp{sys.version_info.major}{sys.version_info.minor}"
    if RDKIT_WHL is not None and py_tag in RDKIT_WHL.name:
        sh(f"{sys.executable} -m pip install -q --no-deps {RDKIT_WHL}", check=False)
    if not v_match(kernel_version("rdkit"), RDKIT_PIN):
        print(f"[INFO] Offline wheel for {py_tag} not found or incompatible; installing from PyPI...", flush=True)
        sh(f"{sys.executable} -m pip install -q rdkit==2026.3.3")

def dist_version(dist):
    # selfies 2.1.2 still reports __version__ == "2.1.1"; the installed distribution's metadata is right
    r = subprocess.run([sys.executable, "-c", f"import importlib.metadata as m; print(m.version('{dist}'))"],
                       capture_output=True, text=True)
    return r.stdout.strip() if r.returncode == 0 else None

if dist_version("selfies") != SELFIES_PIN:
    sh(f"{sys.executable} -m pip install -q selfies=={SELFIES_PIN}")
import rdkit, selfies as sf
assert v_match(rdkit.__version__, RDKIT_PIN), rdkit.__version__
assert dist_version("selfies") == SELFIES_PIN, dist_version("selfies")
GATE_RESULT["env"] = dict(python=sys.version.split()[0], numpy=np.__version__, pandas=pd.__version__,
                          pyarrow=pa.__version__, rdkit=rdkit.__version__, selfies=dist_version("selfies"))
print(GATE_RESULT["env"])
```

```python
NOVEL_DECODER_SRC = '"""specs_novel single source of truth for the SELFIES side and the decoder (plan §4, §5.5).\n\n    SelfiesVocab        token <-> id, the same ids as MassSpecGym\'s SelfiesTokenizer (JSON, no `tokenizers`)\n    TokenTable          per-token grammar facts, precomputed with selfies\' own grammar_rules\n    Tracker             incremental SELFIES derivation state -> legal-token mask (plan §4.1-4.2)\n    reference_state     the slow truth: selfies.decoder(prefix) + RDKit counts (P0.4 parity)\n    walk_truth          mask soundness: every true next token legal, </s> legal at the end (P0.4)\n    prep_structure      stereo/isotope strip -> canonical SMILES -> SELFIES (A9, P0.3)\n\nThe Tracker replays `selfies.decoder._derive_mol_from_symbols` token by token. That function is\nrecursive (a branch derives a sub-chain of Q+1 symbols); here its call stack is an explicit list of\nframes [state, prev_atom, budget]. Every consumed symbol lowers the budget of every open frame, the\nway each caller adds its child\'s n_derived; a frame closes when its budget reaches 0. A frame whose\nstate becomes None (no free valence left) swallows the rest of its budget, and in the main chain that\nmeans every later symbol -- SELFIES silently skips those atoms (A7), so the mask never offers them.\n\nMasks (plan §4.2):\n    pending index symbols      only the 16 INDEX_ALPHABET symbols (the encoder emits nothing else there)\n    atom token, element e      frame active, the atom would be placed, and remaining[e] > 0\n    [Branch*]                  frame state > 1 (otherwise selfies ignores it)\n    [Ring*]                    frame state > 0\n    </s>                       no pending index and every heavy-atom count exactly used\n`unsat()` is rings + pi bonds of the molecule the prefix decodes to, with the ring bonds formed the way\n`_form_rings_bilocally` forms them at the end (orders capped by the free valence left). It feeds the\noptional DBE prune, which is only applied where P0.4 shows it sound (A8).\n\nP3 part (below the P0 part, specs_novel/gate_report.md -> G3):\n    DreamsSelfiesDecoder  memory = per-spectrum tokens (+ adduct, + CE bin) plus a formula token,\n                          no positional encoding on the memory, tied in/out token embeddings\n    search_prefix_seeded  arm A: masked expansion to depth L_SEED, then sampled rollouts per prefix\n    search_masked_beam    arm B (alpha = 0) and arm C (alpha > 0, cross-beam scaffold penalty)\n    select_top25          "maxmin" (diversity, what G3 is judged on) or "logp"\n    max_tanimoto          the G3 metric, on MassSpecGym\'s fingerprint\n\nNeeds numpy + selfies (pinned) for the P0 part; RDKit for reference_state / prep_structure / ik14 /\nthe selection helpers; torch for the P3 part.\n"""\nimport json\nimport re\n\nimport numpy as np\nimport selfies as sf\nfrom selfies.constants import INDEX_CODE\nfrom selfies.grammar_rules import (get_index_from_selfies, next_atom_state, next_branch_state,\n                                   next_ring_state, process_atom_symbol, process_branch_symbol,\n                                   process_ring_symbol)\n\nPAD, SOS, EOS, UNK = "<pad>", "<s>", "</s>", "<unk>"\nALLOWED_ELEMENTS = ("C", "H", "N", "O", "P", "S", "F", "Cl", "Br", "I")\nDBE_SAFE_ELEMENTS = frozenset(("C", "H", "N", "O", "F", "Cl", "Br", "I"))   # no S / P (plan §4.2)\nK_SPECIAL, K_ATOM, K_BRANCH, K_RING, K_OTHER = 0, 1, 2, 3, 4\n_FORMULA_RE = re.compile(r"([A-Z][a-z]?)(\\d*)")\n\n\ndef check_constraints():\n    """The token table and the tracker assume selfies\' default semantic constraints."""\n    assert sf.get_semantic_constraints() == sf.get_preset_constraints("default"), \\\n        "selfies semantic constraints are not the default preset"\n\n\n# --------------------------------------------------------------------------------------------- #\n# Vocabulary\n# --------------------------------------------------------------------------------------------- #\n\ndef vocab_from_alphabet(alphabet):\n    """SelfiesTokenizer\'s id layout (massspecgym/models/tokenizers.py:79-85 + add_special_tokens):\n    sorted alphabet 0..n-1, <unk> n, then <pad> <s> </s> as the next added tokens."""\n    alphabet = sorted(set(alphabet) - {PAD, SOS, EOS, UNK})\n    tok2id = {s: i for i, s in enumerate(alphabet)}\n    tok2id[UNK] = len(tok2id)\n    for s in (PAD, SOS, EOS):\n        tok2id[s] = len(tok2id)\n    return tok2id\n\n\nclass SelfiesVocab:\n    def __init__(self, tok2id):\n        ids = sorted(tok2id.values())\n        assert ids == list(range(len(ids))), "token ids must be 0..n-1"\n        for s in (PAD, SOS, EOS, UNK):\n            assert s in tok2id, f"special token {s} missing"\n        self.tok2id = dict(tok2id)\n        self.id2tok = [None] * len(ids)\n        for s, i in self.tok2id.items():\n            self.id2tok[i] = s\n        self.pad_id, self.sos_id, self.eos_id, self.unk_id = (self.tok2id[s] for s in (PAD, SOS, EOS, UNK))\n\n    def __len__(self):\n        return len(self.id2tok)\n\n    @classmethod\n    def from_alphabet(cls, alphabet):\n        return cls(vocab_from_alphabet(alphabet))\n\n    @classmethod\n    def from_json(cls, path):\n        with open(path, encoding="utf-8") as f:\n            return cls(json.load(f))\n\n    def to_json(self, path):\n        with open(path, "w", encoding="utf-8") as f:\n            json.dump(self.tok2id, f, sort_keys=True, indent=0)\n\n    def encode(self, selfies_str):\n        """SELFIES string -> ids, no <s>/</s>."""\n        return [self.tok2id.get(t, self.unk_id) for t in sf.split_selfies(selfies_str)]\n\n    def decode(self, ids):\n        """ids -> SELFIES string; special tokens dropped."""\n        return "".join(self.id2tok[i] for i in ids if i not in (self.pad_id, self.sos_id, self.eos_id,\n                                                                self.unk_id))\n\n\n# --------------------------------------------------------------------------------------------- #\n# Token table\n# --------------------------------------------------------------------------------------------- #\n\nclass TokenTable:\n    """Per-token facts over a vocabulary, from selfies\' own grammar_rules (so they follow the pin)."""\n\n    def __init__(self, vocab):\n        check_constraints()\n        n = len(vocab)\n        self.vocab = vocab\n        self.kind = np.full(n, K_OTHER, np.int8)\n        self.elem = np.full(n, -1, np.int16)           # index into self.elements\n        self.bond = np.zeros(n, np.int8)               # requested bond order (atom / branch / ring type)\n        self.cap = np.zeros(n, np.int8)                # atom bonding capacity (after explicit H, charge)\n        self.nidx = np.zeros(n, np.int8)               # index symbols read after a branch / ring\n        self.is_index = np.zeros(n, bool)              # one of the 16 INDEX_ALPHABET symbols\n        self.index_code = np.zeros(n, np.int16)\n        elements = []\n        for i, t in enumerate(vocab.id2tok):\n            if t in (PAD, SOS, EOS, UNK):\n                self.kind[i] = K_SPECIAL\n                continue\n            if t in INDEX_CODE:\n                self.is_index[i] = True\n                self.index_code[i] = INDEX_CODE[t]\n            if "ch" == t[-4:-2]:                       # the same dispatch order as the decoder\n                out = process_branch_symbol(t)\n                if out is not None:\n                    self.kind[i], self.bond[i], self.nidx[i] = K_BRANCH, out[0], out[1]\n            elif "ng" == t[-4:-2]:\n                out = process_ring_symbol(t)\n                if out is not None:\n                    self.kind[i], self.bond[i], self.nidx[i] = K_RING, out[0], out[1]\n            elif "eps" in t:\n                pass\n            else:\n                out = process_atom_symbol(t)\n                if out is not None:\n                    (bond, _), atom = out\n                    if atom.element not in elements:\n                        elements.append(atom.element)\n                    self.kind[i] = K_ATOM\n                    self.elem[i] = elements.index(atom.element)\n                    self.bond[i] = bond\n                    self.cap[i] = atom.bonding_capacity\n        self.elements = tuple(elements)\n        self.elem_ix = {e: j for j, e in enumerate(self.elements)}\n        self.h_ix = self.elem_ix.get("H", -1)\n        self.is_atom = self.kind == K_ATOM\n        self.is_branch = self.kind == K_BRANCH\n        self.is_ring = self.kind == K_RING\n        self.atom_cap_pos = self.is_atom & (self.cap > 0)\n        self.elem_safe = np.maximum(self.elem, 0)\n\n    def counts_vector(self, counts):\n        """{element: count} -> vector over self.elements, plus the count of elements the vocab lacks."""\n        v = np.zeros(len(self.elements), np.int32)\n        missing = 0\n        for e, c in counts.items():\n            if e in self.elem_ix:\n                v[self.elem_ix[e]] = c\n            elif e != "H":\n                missing += c\n        return v, missing\n\n\n# --------------------------------------------------------------------------------------------- #\n# Tracker\n# --------------------------------------------------------------------------------------------- #\n\nINF = float("inf")\n\n\nclass Tracker:\n    """Persistent SELFIES derivation state. `push` returns a new Tracker; the old one is unchanged.\n\n    formula: {element: count} incl. H, or None (then only counting, no count masks)."""\n\n    __slots__ = ("tt", "frames", "pending", "caps", "bc", "bonds", "rings", "counts", "target",\n                 "missing", "n_tokens", "broken")\n\n    def __init__(self, table, formula=None):\n        self.tt = table\n        self.frames = [[0, None, INF]]          # [state (int | None), prev atom index, budget]\n        self.pending = None                     # [kind, n_left, symbols, payload]\n        self.caps = []                          # bonding capacity per placed atom\n        self.bc = []                            # bond order sum per atom (chain bonds only)\n        self.bonds = {}                         # (a, b) a < b -> order (chain bonds only)\n        self.rings = []                         # (left atom, right atom, order), formed at the end\n        self.counts = np.zeros(len(table.elements), np.int32)\n        if formula is None:\n            self.target, self.missing = None, 0\n        else:\n            self.target, self.missing = table.counts_vector(formula)\n        self.n_tokens = 0\n        self.broken = False\n\n    def _copy(self):\n        t = Tracker.__new__(Tracker)\n        t.tt = self.tt\n        t.frames = [f[:] for f in self.frames]\n        t.pending = None if self.pending is None else [self.pending[0], self.pending[1],\n                                                       self.pending[2][:], self.pending[3]]\n        t.caps, t.bc, t.bonds, t.rings = self.caps[:], self.bc[:], dict(self.bonds), self.rings[:]\n        t.counts = self.counts.copy()\n        t.target, t.missing, t.n_tokens, t.broken = self.target, self.missing, self.n_tokens, self.broken\n        return t\n\n    # -- derivation -------------------------------------------------------------------------- #\n\n    def _consume(self):\n        for f in self.frames:\n            f[2] -= 1\n\n    def _close_exhausted(self):\n        while len(self.frames) > 1 and self.frames[-1][2] <= 0:\n            self.frames.pop()\n\n    def _add_atom(self, cap):\n        self.caps.append(int(cap))\n        self.bc.append(0)\n        return len(self.caps) - 1\n\n    def push(self, tok_id):\n        t = self._copy()\n        t._apply(int(tok_id))\n        return t\n\n    def _apply(self, i):\n        tt = self.tt\n        self.n_tokens += 1\n        self._consume()\n        if self.pending is not None:                      # an index symbol of a branch / ring\n            p = self.pending\n            p[2].append(tt.vocab.id2tok[i])\n            p[1] -= 1\n            if p[1] == 0:\n                self.pending = None\n                q = get_index_from_selfies(*p[2])\n                f = self.frames[-1]\n                if p[0] == K_BRANCH:\n                    self.frames.append([p[3], f[1], q + 1])\n                else:\n                    lidx = max(0, f[1] - (q + 1))\n                    self.rings.append((lidx, f[1], p[3]))\n            self._close_exhausted()\n            return\n        f = self.frames[-1]\n        state = f[0]\n        kind = tt.kind[i]\n        if state is None or kind in (K_SPECIAL, K_OTHER):\n            if kind == K_OTHER and state is not None and "eps" in tt.vocab.id2tok[i]:\n                f[0] = 0 if state == 0 else None\n            elif kind == K_SPECIAL:\n                self.broken = True                        # specials never occur inside a SELFIES string\n        elif kind == K_BRANCH:\n            if state > 1:\n                binit, f[0] = next_branch_state(int(tt.bond[i]), state)\n                self.pending = [K_BRANCH, int(tt.nidx[i]), [], binit]\n        elif kind == K_RING:\n            if state != 0:\n                order, f[0] = next_ring_state(int(tt.bond[i]), state)\n                self.pending = [K_RING, int(tt.nidx[i]), [], order]\n        else:                                             # atom\n            cap = int(tt.cap[i])\n            order, nxt = next_atom_state(int(tt.bond[i]), cap, state)\n            if order == 0:\n                if state == 0:\n                    f[1] = self._add_atom(cap)\n                    self.counts[tt.elem[i]] += 1\n                else:\n                    self.broken = True                    # selfies would bond from an unplaced atom\n            else:\n                a = self._add_atom(cap)\n                self.counts[tt.elem[i]] += 1\n                src = f[1]\n                self.bonds[(src, a)] = order\n                self.bc[src] += order\n                self.bc[a] += order\n                f[1] = a\n            f[0] = nxt\n        self._close_exhausted()\n\n    # -- queries ----------------------------------------------------------------------------- #\n\n    def remaining(self):\n        return None if self.target is None else self.target - self.counts\n\n    def heavy_done(self):\n        r = self.remaining()\n        if r is None:\n            return True\n        r = r.copy()\n        if self.tt.h_ix >= 0:\n            r[self.tt.h_ix] = 0\n        return self.missing == 0 and not r.any()\n\n    def eos_legal(self):\n        return (not self.broken and self.pending is None and len(self.caps) > 0 and self.heavy_done())\n\n    def legal_mask(self):\n        """numpy bool over the vocabulary."""\n        tt = self.tt\n        n = len(tt.kind)\n        if self.broken:\n            return np.zeros(n, bool)\n        if self.pending is not None:\n            return tt.is_index.copy()\n        state = self.frames[-1][0]\n        mask = np.zeros(n, bool)\n        if state is not None:\n            placed = tt.is_atom if state == 0 else tt.atom_cap_pos\n            if self.target is not None:\n                placed = placed & (self.remaining()[tt.elem_safe] > 0)\n            mask |= placed\n            if state > 1:\n                mask |= tt.is_branch\n            if state > 0:\n                mask |= tt.is_ring\n        if self.eos_legal():\n            mask[tt.vocab.eos_id] = True\n        return mask\n\n    def unsat(self):\n        """Rings + pi bonds of the molecule this prefix decodes to (selfies\' _form_rings_bilocally)."""\n        n = len(self.caps)\n        if n == 0:\n            return 0\n        bc, bonds = self.bc[:], dict(self.bonds)\n        for l, r, order in self.rings:\n            if l == r:\n                continue\n            lfree, rfree = self.caps[l] - bc[l], self.caps[r] - bc[r]\n            if lfree <= 0 or rfree <= 0:\n                continue\n            order = min(order, lfree, rfree)\n            key = (min(l, r), max(l, r))\n            if key in bonds:\n                new = min(order + bonds[key], 3)\n                bc[l] += new - bonds[key]\n                bc[r] += new - bonds[key]\n                bonds[key] = new\n            else:\n                bonds[key] = order\n                bc[l] += order\n                bc[r] += order\n        return sum(bonds.values()) - n + 1\n\n    def element_counts(self):\n        return {e: int(c) for e, c in zip(self.tt.elements, self.counts) if c}\n\n\n# --------------------------------------------------------------------------------------------- #\n# Formulae\n# --------------------------------------------------------------------------------------------- #\n\ndef parse_formula(formula):\n    counts = {}\n    for e, c in _FORMULA_RE.findall(formula):\n        counts[e] = counts.get(e, 0) + (int(c) if c else 1)\n    return counts\n\n\ndef dbe(counts):\n    """Rings + pi bonds implied by a formula with standard valences (C4 N3 O2 X1 H1)."""\n    x = sum(counts.get(e, 0) for e in ("F", "Cl", "Br", "I"))\n    return 1 + counts.get("C", 0) - (counts.get("H", 0) + x) / 2 + counts.get("N", 0) / 2\n\n\ndef dbe_prune_applies(counts):\n    return set(e for e, c in counts.items() if c) <= DBE_SAFE_ELEMENTS\n\n\n# --------------------------------------------------------------------------------------------- #\n# RDKit side: structure prep, ik14, reference tracker\n# --------------------------------------------------------------------------------------------- #\n\ndef _chem():\n    from rdkit import Chem, RDLogger\n    RDLogger.DisableLog("rdApp.*")\n    return Chem\n\n\ndef mol_formula(mol):\n    """{element: count} incl. total H (implicit + explicit) of an RDKit mol."""\n    counts = {}\n    for a in mol.GetAtoms():\n        s = a.GetSymbol()\n        counts[s] = counts.get(s, 0) + 1\n        h = a.GetTotalNumHs()\n        if h:\n            counts["H"] = counts.get("H", 0) + h\n    return counts\n\n\ndef ik14(smiles):\n    Chem = _chem()\n    m = Chem.MolFromSmiles(smiles) if smiles else None\n    if m is None:\n        return None\n    k = Chem.MolToInchiKey(m)\n    return k[:14] if k else None\n\n\ndef prep_structure(smiles, elements=ALLOWED_ELEMENTS):\n    """-> (status, canonical_smiles, selfies, formula). status \'ok\' or the reason it is excluded:\n    \'parse\', \'multi\' (more than one component), \'element\', \'encode\' (selfies constraints)."""\n    Chem = _chem()\n    m = Chem.MolFromSmiles(smiles) if smiles else None\n    if m is None:\n        return "parse", None, None, None\n    if len(Chem.GetMolFrags(m)) > 1:\n        return "multi", None, None, None\n    if any(a.GetSymbol() not in elements for a in m.GetAtoms()):\n        return "element", None, None, None\n    Chem.RemoveStereochemistry(m)\n    for a in m.GetAtoms():\n        a.SetIsotope(0)\n    can = Chem.MolToSmiles(m)\n    m = Chem.MolFromSmiles(can)\n    try:\n        s = sf.encoder(can, strict=True)\n    except Exception:\n        return "encode", can, None, None\n    if s is None:\n        return "encode", can, None, None\n    return "ok", can, s, mol_formula(m)\n\n\ndef reference_state(tokens):\n    """Slow truth for a token prefix: (element counts incl. explicit H atoms, rings + pi bonds)."""\n    Chem = _chem()\n    smi = sf.decoder("".join(tokens))\n    if not smi:\n        return {}, 0\n    m = Chem.MolFromSmiles(smi, sanitize=False)\n    counts = {}\n    for a in m.GetAtoms():\n        counts[a.GetSymbol()] = counts.get(a.GetSymbol(), 0) + 1\n    order = sum(int(b.GetBondTypeAsDouble()) for b in m.GetBonds())\n    n_comp = len(Chem.GetMolFrags(m, sanitizeFrags=False))\n    return counts, order - m.GetNumAtoms() + n_comp\n\n\ndef tracker_parity(table, tokens):\n    """Fast vs reference at every prefix. -> list of (prefix length, what, fast, reference)."""\n    vocab = table.vocab\n    t = Tracker(table)\n    bad = []\n    for k, tok in enumerate(tokens, 1):\n        t = t.push(vocab.tok2id[tok])\n        ref_counts, ref_u = reference_state(tokens[:k])\n        if t.element_counts() != ref_counts:\n            bad.append((k, "counts", t.element_counts(), ref_counts))\n        if t.pending is None and t.unsat() != ref_u:\n            bad.append((k, "unsat", t.unsat(), ref_u))\n        if t.broken:\n            bad.append((k, "broken", None, None))\n    return bad\n\n\ndef walk_truth(table, tokens, formula, max_len=None, dbe_prune=False):\n    """Mask soundness for one truth. -> list of (position, rule, token) violations.\n\n    rules: \'count\' (atom masked, element exhausted), \'grammar\' (masked for any other reason),\n    \'eos\' (</s> not legal after the last token), \'length\' (> max_len), \'dbe\' (unsat above DBE)."""\n    vocab = table.vocab\n    t = Tracker(table, formula)\n    limit = dbe(formula) if dbe_prune and dbe_prune_applies(formula) else None\n    bad = []\n    for k, tok in enumerate(tokens):\n        i = vocab.tok2id.get(tok, vocab.unk_id)\n        if not t.legal_mask()[i]:\n            r = t.remaining()\n            count_rule = (table.kind[i] == K_ATOM and t.pending is None and r is not None\n                          and r[table.elem_safe[i]] <= 0)\n            bad.append((k, "count" if count_rule else "grammar", tok))\n        t = t.push(i)\n        if limit is not None and t.pending is None and t.unsat() > limit + 1e-9:\n            bad.append((k, "dbe", tok))\n    if not t.eos_legal():\n        bad.append((len(tokens), "eos", None))\n    if max_len is not None and len(tokens) > max_len:\n        bad.append((len(tokens), "length", None))\n    return bad\n\n\n# --------------------------------------------------------------------------------------------- #\n# P3: memory, decoder, batched constrained search (torch)\n#\n# Everything below is the P3 part (specs_novel/gate_report.md -> G3). The P0 part above stays\n# torch-free; torch is imported here because every consumer of this part needs it anyway (Colab\n# training, the Kaggle worker, the local tests).\n# --------------------------------------------------------------------------------------------- #\n\nimport math\n\nimport torch\nimport torch.nn as nn\nimport torch.nn.functional as F\n\n# must equal simsim_common.TEST_ADDUCTS -- test_novel_p3.check_adducts asserts it\nTEST_ADDUCTS = ("[M+H]+", "[M+NH4]+", "[M-H2O+H]+", "[M-2H2O+H]+", "[M+Na]+", "[M+K]+",\n                "[M-H]-", "[M-H2O-H]-", "[M+CH2O2-H]-", "[M+Cl]-")\nCE_BINS = (15.0, 23.0, 36.0, 44.0)       # P0.2 quantiles; 5 intervals + "unknown" = 6 rows\nN_CE_BINS = len(CE_BINS) + 2             # digitize gives 0..4, and 5 is "unknown"\nCE_UNKNOWN = len(CE_BINS) + 1\nFORMULA_SCALE = 10.0                     # counts are divided by this before the formula MLP\n\n\ndef ce_bin(ce, bins=CE_BINS):\n    """Collision energy in eV -> bin index. None / NaN / <= 0 -> CE_UNKNOWN."""\n    if ce is None:\n        return CE_UNKNOWN\n    try:\n        v = float(ce)\n    except (TypeError, ValueError):\n        return CE_UNKNOWN\n    if not np.isfinite(v) or v <= 0.0:\n        return CE_UNKNOWN\n    return int(np.digitize(v, bins))\n\n\ndef formula_vector(counts, elements=ALLOWED_ELEMENTS):\n    """{element: count} -> float32 vector over `elements`, hydrogen INCLUDED.\n\n    Deliberately not TokenTable.counts_vector: that one drops H (the vocabulary has no free H token),\n    but the formula token conditions on the full formula, H and all."""\n    v = np.zeros(len(elements), np.float32)\n    ix = {e: i for i, e in enumerate(elements)}\n    for e, c in counts.items():\n        if e in ix:\n            v[ix[e]] = c\n    return v\n\n\nclass DreamsSelfiesDecoder(nn.Module):\n    """SELFIES decoder over frozen DreaMS embeddings (plan section 3.5).\n\n    memory = [W_e.emb_i + E_adduct[a_i] + E_ce[c_i]]_i  (+)  [MLP(formula_counts)]\n\n    There is no positional encoding on the memory: the spectra of a molecule are a set, and\n    test_novel_p3 asserts the permutation invariance. Input and output token embeddings are tied.\n    """\n\n    def __init__(self, n_vocab, *, pad_id, sos_id, eos_id, d_model=512, n_layers=8, n_heads=8,\n                 d_ff=2048, dropout=0.1, max_len=171, n_elements=len(ALLOWED_ELEMENTS),\n                 n_adducts=len(TEST_ADDUCTS), n_ce_bins=N_CE_BINS, d_emb=1024):\n        super().__init__()\n        self.hparams = dict(n_vocab=n_vocab, pad_id=pad_id, sos_id=sos_id, eos_id=eos_id,\n                            d_model=d_model, n_layers=n_layers, n_heads=n_heads, d_ff=d_ff,\n                            dropout=dropout, max_len=max_len, n_elements=n_elements,\n                            n_adducts=n_adducts, n_ce_bins=n_ce_bins, d_emb=d_emb)\n        self.pad_id, self.sos_id, self.eos_id = pad_id, sos_id, eos_id\n        self.d_model, self.max_len = d_model, max_len\n\n        # -- memory side --\n        self.emb_proj = nn.Linear(d_emb, d_model)\n        self.adduct_emb = nn.Embedding(n_adducts, d_model)\n        self.ce_emb = nn.Embedding(n_ce_bins, d_model)\n        self.formula_mlp = nn.Sequential(nn.Linear(n_elements, d_model), nn.GELU(),\n                                         nn.Linear(d_model, d_model))\n        self.mem_norm = nn.LayerNorm(d_model)\n\n        # -- token side --\n        self.tok_emb = nn.Embedding(n_vocab, d_model, padding_idx=pad_id)\n        self.pos_emb = nn.Embedding(max_len + 2, d_model)\n        self.drop = nn.Dropout(dropout)\n        layer = nn.TransformerDecoderLayer(d_model, n_heads, d_ff, dropout, activation="gelu",\n                                           batch_first=True, norm_first=True)\n        self.dec = nn.TransformerDecoder(layer, n_layers, norm=nn.LayerNorm(d_model))\n        self.out_bias = nn.Parameter(torch.zeros(n_vocab))\n        self.reset_parameters()\n\n    def reset_parameters(self):\n        nn.init.normal_(self.tok_emb.weight, std=0.02)\n        with torch.no_grad():\n            self.tok_emb.weight[self.pad_id].zero_()\n        nn.init.normal_(self.pos_emb.weight, std=0.02)\n\n    # -- memory ------------------------------------------------------------------------------- #\n\n    def build_memory(self, embs, adduct_ids, ce_ids, formula_counts, spec_mask=None):\n        """embs (B, K, d_emb) float, adduct_ids / ce_ids (B, K) long, formula_counts (B, E) float,\n        spec_mask (B, K) bool with True = a real spectrum.\n\n        -> memory (B, K+1, D), mem_pad (B, K+1) bool with True = ignore (nn.Transformer convention).\n        The formula token is last and is never padded."""\n        if embs.dim() == 2:                       # a single molecule, (K, d_emb)\n            embs, adduct_ids, ce_ids = embs[None], adduct_ids[None], ce_ids[None]\n            formula_counts = formula_counts[None]\n            spec_mask = None if spec_mask is None else spec_mask[None]\n        b, k = embs.shape[0], embs.shape[1]\n        if spec_mask is None:\n            spec_mask = torch.ones(b, k, dtype=torch.bool, device=embs.device)\n        m_spec = self.emb_proj(embs) + self.adduct_emb(adduct_ids) + self.ce_emb(ce_ids)\n        m_form = self.formula_mlp(formula_counts / FORMULA_SCALE).unsqueeze(1)\n        memory = self.mem_norm(torch.cat([m_spec, m_form], dim=1))\n        mem_pad = torch.cat([~spec_mask, torch.zeros(b, 1, dtype=torch.bool, device=embs.device)],\n                            dim=1)\n        # zero the padded spectrum tokens so they cannot leak through a fully-masked attention row\n        memory = memory.masked_fill(mem_pad.unsqueeze(-1), 0.0)\n        return memory, mem_pad\n\n    # -- decoding ----------------------------------------------------------------------------- #\n\n    def forward(self, tok_ids, memory, mem_pad=None):\n        """tok_ids (B, L) long, already prefixed with <s>. -> logits (B, L, V)."""\n        b, l = tok_ids.shape\n        assert l <= self.max_len + 2, f"sequence length {l} exceeds max_len + 2"\n        pos = torch.arange(l, device=tok_ids.device)\n        x = self.drop(self.tok_emb(tok_ids) * math.sqrt(self.d_model) + self.pos_emb(pos))\n        causal = torch.triu(torch.ones(l, l, dtype=torch.bool, device=tok_ids.device), diagonal=1)\n        h = self.dec(x, memory, tgt_mask=causal, memory_key_padding_mask=mem_pad)\n        return h @ self.tok_emb.weight.T + self.out_bias\n\n    def next_logits(self, tok_ids, memory, mem_pad=None):\n        """Logits for the token after the last position of `tok_ids`. (B, V).\n\n        Recomputes the whole prefix each call: no KV cache. At this model size that is the simpler\n        correct thing; a cache is a P4 optimisation (search_plan.md v1.1 section 0.5)."""\n        return self.forward(tok_ids, memory, mem_pad)[:, -1, :]\n\n\n# --------------------------------------------------------------------------------------------- #\n# Search: shared scaffolding\n# --------------------------------------------------------------------------------------------- #\n\nNEG_INF = -1e9\n\n\ndef _expand_memory(memory, mem_pad, n):\n    """A single molecule\'s memory -> n identical rows, so a whole beam is one forward pass."""\n    if memory.shape[0] == n:\n        return memory, mem_pad\n    assert memory.shape[0] == 1, f"expected a single-molecule memory, got {memory.shape[0]}"\n    pad = None if mem_pad is None else mem_pad.expand(n, -1)\n    return memory.expand(n, -1, -1), pad\n\n\ndef _masks_to_tensor(trackers, device):\n    return torch.from_numpy(np.stack([t.legal_mask() for t in trackers])).to(device)\n\n\ndef _logprobs(model, tok_ids, memory, mem_pad, trackers):\n    """Masked log-probabilities of the next token. Illegal tokens get NEG_INF, so a beam can never\n    step outside the formula. Rows with no legal token at all are reported as `dead`."""\n    logits = model.next_logits(tok_ids, memory, mem_pad).float()\n    legal = _masks_to_tensor(trackers, logits.device)\n    dead = ~legal.any(dim=1)\n    logits = logits.masked_fill(~legal, NEG_INF)\n    return torch.log_softmax(logits, dim=-1), legal, dead\n\n\ndef _leaf(vocab, ids, target_counts):\n    """A finished token sequence -> (smiles, ik14), or None if it fails the plan section 3.6 filter.\n\n    The tracker already guarantees the heavy-atom counts; what is checked here is everything it\n    cannot know: that the string decodes and sanitises, that it is one component, and that the\n    FULL formula including hydrogen matches. Hydrogen is only fixed once SELFIES has filled the\n    valences, which is why this check does real work."""\n    Chem = _chem()\n    smi = sf.decoder(vocab.decode(ids))\n    if not smi:\n        return None\n    m = Chem.MolFromSmiles(smi)\n    if m is None or len(Chem.GetMolFrags(m)) > 1:\n        return None\n    if mol_formula(m) != target_counts:\n        return None\n    can = Chem.MolToSmiles(m)\n    k = Chem.MolToInchiKey(m)\n    return (can, k[:14]) if k else None\n\n\ndef _collect(out, vocab, ids, logp, target_counts):\n    """Add a finished sequence to `out`, an ik14 -> (smiles, logp) dict keeping the best logp."""\n    leaf = _leaf(vocab, ids, target_counts)\n    if leaf is None:\n        return False\n    smi, key = leaf\n    prev = out.get(key)\n    if prev is None or logp > prev[1]:\n        out[key] = (smi, logp)\n    return True\n\n\ndef _prep_search(model, memory, mem_pad, formula, table):\n    """Shared entry: canonical target counts, the root tracker, and the <s> row."""\n    counts = parse_formula(formula) if isinstance(formula, str) else dict(formula)\n    counts = {e: int(c) for e, c in counts.items() if c}\n    root = Tracker(table, counts)\n    assert root.missing == 0, f"formula {formula} has elements outside the vocabulary"\n    device = memory.device\n    sos = torch.full((1, 1), model.sos_id, dtype=torch.long, device=device)\n    return counts, root, sos\n\n\n# --------------------------------------------------------------------------------------------- #\n# Arm A: prefix-seeded batched sampling  (search_plan.md v1.1 Method 1, corrected)\n# --------------------------------------------------------------------------------------------- #\n\ndef _seed_prefixes(model, memory, mem_pad, table, root, sos, l_seed, k_roots, width, prefix_cap):\n    """Beam-expand to depth `l_seed`, then keep `k_roots` prefixes with a spread of scaffolds.\n\n    Seeding at depth 1 would be pointless: at derivation state 0 the mask admits only atom tokens\n    ([Branch*] needs state > 1, [Ring*] needs state > 0) and next_atom_state returns order 0 for\n    every bond prefix, so [C] / [=C] / [#C] are degenerate. The distinct roots at depth 1 are just\n    the elements of the formula. Ring and branch structure -- the scaffold -- first appears a few\n    tokens in, which is why the default is l_seed = 4.\n\n    `prefix_cap` bounds how many kept prefixes may share their first two tokens, so the K roots\n    cannot all collapse into one lineage. The cap is relaxed if too few prefixes survive it."""\n    beam = [(sos, root, 0.0)]                      # (ids (1, L), tracker, logp)\n    for _ in range(l_seed):\n        ids = torch.cat([b[0] for b in beam], dim=0)\n        trackers = [b[1] for b in beam]\n        mem, pad = _expand_memory(memory, mem_pad, len(beam))\n        lp, legal, dead = _logprobs(model, ids, mem, pad, trackers)\n        cand = []\n        for i, (bi, t, base) in enumerate(beam):\n            if dead[i]:\n                continue\n            for tok in torch.nonzero(legal[i], as_tuple=False).flatten().tolist():\n                if tok == model.eos_id:            # a complete molecule this early is not a seed\n                    continue\n                cand.append((base + float(lp[i, tok]), i, tok))\n        if not cand:\n            break\n        cand.sort(key=lambda c: -c[0])\n        nxt = []\n        for score, i, tok in cand[:width]:\n            bi, t, _ = beam[i]\n            step = torch.tensor([[tok]], dtype=torch.long, device=bi.device)\n            nxt.append((torch.cat([bi, step], dim=1), t.push(tok), score))\n        beam = nxt\n    beam.sort(key=lambda b: -b[2])\n    for cap in (prefix_cap, len(beam)):            # relax the spread rule rather than return fewer\n        seen, kept = {}, []\n        for b in beam:\n            sig = tuple(b[0][0, 1:3].tolist())\n            if seen.get(sig, 0) >= cap:\n                continue\n            seen[sig] = seen.get(sig, 0) + 1\n            kept.append(b)\n            if len(kept) == k_roots:\n                break\n        if len(kept) >= min(k_roots, len(beam)):\n            return kept\n    return kept\n\n\ndef _nucleus(lp, top_p):\n    """Zero out the tail of each row beyond cumulative probability `top_p` (log space in, log out)."""\n    srt, idx = torch.sort(lp, dim=-1, descending=True)\n    keep = torch.cumsum(srt.exp(), dim=-1) - srt.exp() < top_p\n    keep[:, 0] = True                              # never drop the mode\n    drop = torch.zeros_like(keep).scatter(1, idx, ~keep)\n    return torch.log_softmax(lp.masked_fill(drop, NEG_INF), dim=-1)\n\n\n@torch.no_grad()\ndef search_prefix_seeded(model, memory, formula, table, *, mem_pad=None, l_seed=4, k_roots=8,\n                         m_rollout=32, temperature=0.7, top_p=0.90, max_len=None, seed=None):\n    """P3 bake-off arm A. -> [(smiles, logp)], ik14-deduped, best logp per ik14."""\n    model.eval()\n    max_len = max_len or model.max_len\n    counts, root, sos = _prep_search(model, memory, mem_pad, formula, table)\n    gen = None\n    if seed is not None:\n        gen = torch.Generator(device=memory.device).manual_seed(int(seed))\n    prefixes = _seed_prefixes(model, memory, mem_pad, table, root, sos, l_seed, k_roots,\n                              width=max(k_roots * 8, 32), prefix_cap=2)\n    out = {}\n    if not prefixes:\n        return []\n    ids = torch.cat([p[0].repeat(m_rollout, 1) for p in prefixes], dim=0)\n    trackers = [p[1]._copy() for p in prefixes for _ in range(m_rollout)]\n    logp = [p[2] for p in prefixes for _ in range(m_rollout)]\n    alive = list(range(ids.shape[0]))\n    while alive and ids.shape[1] < max_len + 1:\n        act = [trackers[i] for i in alive]\n        mem, pad = _expand_memory(memory, mem_pad, len(alive))\n        lp, legal, dead = _logprobs(model, ids[alive], mem, pad, act)\n        lp = _nucleus(lp / max(temperature, 1e-6), top_p)\n        tok = torch.multinomial(lp.exp(), 1, generator=gen).flatten()\n        step = torch.full((ids.shape[0], 1), model.pad_id, dtype=torch.long, device=ids.device)\n        step[alive] = tok[:, None]\n        ids = torch.cat([ids, step], dim=1)\n        nxt = []\n        for j, i in enumerate(alive):\n            if dead[j]:\n                continue\n            t = int(tok[j])\n            logp[i] += float(lp[j, t])\n            if t == model.eos_id:\n                _collect(out, table.vocab, ids[i, 1:-1].tolist(), logp[i], counts)\n                continue\n            trackers[i] = trackers[i].push(t)\n            nxt.append(i)\n        alive = nxt\n    return sorted(((s, lg) for s, lg in out.values()), key=lambda c: -c[1])\n\n\n# --------------------------------------------------------------------------------------------- #\n# Arms B and C: masked beam search, plain and cross-beam diversified\n# --------------------------------------------------------------------------------------------- #\n\n@torch.no_grad()\ndef search_masked_beam(model, memory, formula, table, *, mem_pad=None, beam=256, max_len=None,\n                       alpha=0.0, max_leaves=None):\n    """P3 bake-off arm B (alpha = 0) and arm C (alpha > 0). -> [(smiles, logp)], ik14-deduped.\n\n    With alpha > 0 the score used for *selection* is penalised by how often each ring / branch token\n    was taken across the previous step\'s beams, which spreads the beams over scaffold topologies\n    instead of letting them collapse onto one core. The returned logp is always the UNPENALISED\n    cumulative log-probability, because P4 orders novel candidates on it.\n\n    This is the non-circular reading of search_plan.md Method 3: the penalty uses the previous\n    step\'s counts, since a penalty defined by the current step\'s own choices is not well posed."""\n    model.eval()\n    max_len = max_len or model.max_len\n    max_leaves = beam * 4 if max_leaves is None else max_leaves\n    counts, root, sos = _prep_search(model, memory, mem_pad, formula, table)\n    scaffold = torch.from_numpy(table.is_branch | table.is_ring).to(memory.device)\n    live = [(sos, root, 0.0)]                      # (ids, tracker, unpenalised logp)\n    out = {}\n    prev = torch.zeros(len(table.kind), device=memory.device)\n    for _ in range(max_len + 1):\n        if not live:\n            break\n        ids = torch.cat([b[0] for b in live], dim=0)\n        trackers = [b[1] for b in live]\n        mem, pad = _expand_memory(memory, mem_pad, len(live))\n        lp, legal, dead = _logprobs(model, ids, mem, pad, trackers)\n        pen = (alpha * (prev / max(len(live), 1)) * scaffold)[None, :] if alpha else 0.0\n        cand = []\n        for i, (bi, t, base) in enumerate(live):\n            if dead[i]:\n                continue\n            for tok in torch.nonzero(legal[i], as_tuple=False).flatten().tolist():\n                real = base + float(lp[i, tok])\n                rank = real - (float(pen[0, tok]) if alpha else 0.0)\n                cand.append((rank, real, i, tok))\n        if not cand:\n            break\n        cand.sort(key=lambda c: -c[0])\n        nxt, taken = [], torch.zeros_like(prev)\n        for _rank, real, i, tok in cand:\n            full = len(nxt) >= beam\n            if full and len(out) >= max_leaves:\n                break\n            bi, t, _ = live[i]\n            if tok == model.eos_id:\n                if len(out) < max_leaves:\n                    _collect(out, table.vocab, bi[0, 1:].tolist(), real, counts)\n                continue\n            if full:\n                continue\n            step = torch.tensor([[tok]], dtype=torch.long, device=bi.device)\n            nxt.append((torch.cat([bi, step], dim=1), t.push(tok), real))\n            taken[tok] += 1\n        live, prev = nxt, taken\n    return sorted(((s, lg) for s, lg in out.values()), key=lambda c: -c[1])\n\n\ndef search_diverse_beam(model, memory, formula, table, *, mem_pad=None, beam=256, alpha=0.6,\n                        max_len=None, max_leaves=None):\n    """P3 bake-off arm C: search_masked_beam with the cross-beam penalty switched on."""\n    return search_masked_beam(model, memory, formula, table, mem_pad=mem_pad, beam=beam,\n                              max_len=max_len, alpha=alpha, max_leaves=max_leaves)\n\n\nSEARCHES = {"prefix_seeded": search_prefix_seeded,\n            "masked_beam": search_masked_beam,\n            "diverse_beam": search_diverse_beam}\n\n\n# --------------------------------------------------------------------------------------------- #\n# Selecting the 25 that are scored\n# --------------------------------------------------------------------------------------------- #\n\ndef morgan_fps(smiles, radius=2, n_bits=2048):\n    """MassSpecGym\'s fingerprint (utils.morgan_fp): GetMorganFingerprintAsBitVect(r=2, 2048)."""\n    Chem = _chem()\n    from rdkit import DataStructs\n    fps = []\n    for s in smiles:\n        m = Chem.MolFromSmiles(s)\n        fps.append(None if m is None else Chem.GetMorganFingerprintAsBitVect(m, radius, nBits=n_bits))\n    return fps, DataStructs\n\n\ndef max_tanimoto(cands, truth_smiles):\n    """The metric behind G3: the best Tanimoto to the truth over a candidate list. 0.0 if empty."""\n    if not cands:\n        return 0.0\n    fps, DataStructs = morgan_fps([truth_smiles] + [c[0] for c in cands])\n    if fps[0] is None:\n        return 0.0\n    ok = [f for f in fps[1:] if f is not None]\n    if not ok:\n        return 0.0\n    return float(max(DataStructs.BulkTanimotoSimilarity(fps[0], ok)))\n\n\ndef select_top25(cands, mode="maxmin", k=25):\n    """Cut a candidate list to `k`. `cands` is [(smiles, logp)], already ik14-deduped.\n\n    "logp"   -- the most likely k. The honest decoder-likelihood ranking.\n    "maxmin" -- start from the most likely candidate, then repeatedly add whichever candidate is\n                furthest (in Tanimoto) from everything already chosen. G3 scores the *maximum*\n                similarity over the k, so covering chemical space beats stacking near-duplicates\n                of one mode (search_plan.md v1.1 section 0.2).\n    """\n    if mode not in ("logp", "maxmin"):\n        raise ValueError(f"unknown selection mode {mode!r}")\n    ranked = sorted(cands, key=lambda c: -c[1])\n    if mode == "logp" or len(ranked) <= k:\n        return ranked[:k]\n    fps, DataStructs = morgan_fps([c[0] for c in ranked])\n    pool = [i for i, f in enumerate(fps) if f is not None]\n    if len(pool) <= k:\n        return [ranked[i] for i in pool] or ranked[:k]\n    chosen = [pool[0]]                             # pool is logp-sorted, so this is the mode\n    rest = pool[1:]\n    best = list(DataStructs.BulkTanimotoSimilarity(fps[chosen[0]], [fps[i] for i in rest]))\n    while len(chosen) < k and rest:\n        j = min(range(len(rest)), key=lambda x: (best[x], -ranked[rest[x]][1]))\n        chosen.append(rest.pop(j))\n        best.pop(j)\n        if not rest:\n            break\n        sims = DataStructs.BulkTanimotoSimilarity(fps[chosen[-1]], [fps[i] for i in rest])\n        best = [max(b, s) for b, s in zip(best, sims)]\n    return [ranked[i] for i in chosen]\n'
SIMSIM_COMMON_SRC = '"""Shared helpers for the specs_simple_ms_simulator Phase-1 notebooks.\n\nInlined verbatim into the generated notebooks (at the SIMSIM_COMMON placeholder) by build/mk_simple_sim_grid.py,\nand imported directly by build/test_simple_sim_grid.py. Pure numpy / pandas / pyarrow: no RDKit, no\ntorch, no ms-pred. `prep` / `esim` call phase1\'s numba kernels (`_clean`, `entropy_sim`), which the\nnotebooks embed in an earlier cell; the tests inject them.\n\nLifted from build/simple_sim_gate.py (sh, sha1, formula_mass, the POOL build, to_model_ce, pb_sf, the\nworker call) and from phase1 (ADDUCTS physics, the stage-A/B window logic). The gate notebook itself is\nleft untouched.\n"""\nimport hashlib\nimport json\nimport os\nimport pickle\nimport re\nimport subprocess\nimport time\nfrom pathlib import Path\n\nimport numpy as np\nimport pandas as pd\nimport pyarrow.parquet as pq\n\nTEST_ADDUCTS = ["[M+H]+", "[M+NH4]+", "[M-H2O+H]+", "[M-2H2O+H]+", "[M+Na]+", "[M+K]+",\n                "[M-H]-", "[M-H2O-H]-", "[M+CH2O2-H]-", "[M+Cl]-"]\n\n# phase1 CFG values (spectrum cleaning and the candidate window) -- unchanged\nFLOOR, TOPK, POWER, ENTW, TOL = 0.002, 64, 1.0, True, 0.01\nPPM_WIN, PPM_CENTER, PPM_FALLBACK = 8.5, 1.45, 30.0\nPPM_EXPAND = (8.5, 20.0, 45.0, 80.0)\nTOPN = 25\n\n# --- adduct physics: phase1 cell 3, restricted to the ten test adducts (all n = 1, z = 1) -----------\n_M = dict(H=1.00782503207, N=14.0030740048, O=15.9949146196, C=12.0, Na=22.9897692809,\n          K=38.96370668, Cl=34.96885268)\n_E = 0.00054857990\n_PROTON = _M["H"] - _E\n_H2O = 2 * _M["H"] + _M["O"]\n_NH4 = _M["N"] + 4 * _M["H"]\n_FORMATE = _M["C"] + 2 * _M["H"] + 2 * _M["O"]\nADDUCT_DELTA = {\n    "[M+H]+": _PROTON, "[M+NH4]+": _NH4 - _E, "[M+Na]+": _M["Na"] - _E, "[M+K]+": _M["K"] - _E,\n    "[M-H2O+H]+": _PROTON - _H2O, "[M-2H2O+H]+": _PROTON - 2 * _H2O,\n    "[M-H]-": -_PROTON, "[M-H2O-H]-": -_PROTON - _H2O, "[M+CH2O2-H]-": _FORMATE - _PROTON,\n    "[M+Cl]-": _M["Cl"] + _E,\n}\n\n\ndef neutral_mass1(mz, adduct):\n    """ONE precursor m/z -> neutral mass (phase1 neutral_mass, test adducts only; NaN elsewhere).\n\n    Not called `neutral_mass`: the submission notebook also embeds phase1\'s vectorised neutral_mass.\n    """\n    d = ADDUCT_DELTA.get(adduct)\n    return float(mz) - d if d is not None else np.nan\n\n\ndef precursor_of(mass, adduct):\n    """neutral exact mass -> theoretical precursor m/z (the inverse of neutral_mass1)."""\n    return np.asarray(mass, dtype=np.float64) + ADDUCT_DELTA[adduct]\n\n\n# --- label formula -> exact neutral mass (as phase1 / the gate; no RDKit) ---------------------------\n_MONO = {"C": 12.0, "H": 1.0078250319, "N": 14.0030740052, "O": 15.9949146221, "S": 31.97207069,\n         "P": 30.97376151, "F": 18.99840320, "Cl": 34.96885271, "Br": 78.9183376, "I": 126.904468,\n         "Si": 27.9769265327, "Se": 79.9165196, "B": 11.0093055, "Na": 22.98976928, "K": 38.96370649,\n         "Fe": 55.9349375, "D": 2.0141017778, "As": 74.9215965, "Li": 7.0160034, "Sn": 119.9021947}\n_TOK = re.compile(r"([A-Z][a-z]?)(\\d*)")\n\n\ndef formula_mass(f):\n    if not isinstance(f, str):\n        return np.nan\n    s, pos = 0.0, 0\n    for m in _TOK.finditer(f):\n        if m.start() != pos:\n            return np.nan\n        pos = m.end()\n        if m.group(1) not in _MONO:\n            return np.nan\n        s += _MONO[m.group(1)] * (int(m.group(2)) if m.group(2) else 1)\n    return s if pos == len(f) else np.nan\n\n\n# --- shell / files ------------------------------------------------------------------------------------\ndef sh(cmd, timeout=None, env=None):\n    print("$", cmd if len(cmd) < 300 else cmd[:300] + " ...", flush=True)\n    r = subprocess.run(cmd, shell=True, text=True, capture_output=True, timeout=timeout, env=env)\n    if r.stdout.strip():\n        print(r.stdout[-2500:])\n    if r.returncode:\n        print(r.stderr[-4000:])\n        raise RuntimeError(f"command failed ({r.returncode})")\n    return r.stdout\n\n\ndef sha1(path, chunk=1 << 20):\n    h = hashlib.sha1()\n    with open(path, "rb") as f:\n        for b in iter(lambda: f.read(chunk), b""):\n            h.update(b)\n    return h.hexdigest()\n\n\n# --- ICEBERG worker call (the worker runs in its own venv; files only) -------------------------------\ndef to_model_ce(ce_ev, prec, mode):\n    return ce_ev if mode == "ev" else ce_ev * 500.0 / prec     # ms-pred\'s own ev_to_nce\n\n\ndef write_request(req, path):\n    req[["key", "smiles", "adduct", "precursor_mz", "ce", "instrument"]].to_csv(path, sep="\\t", index=False)\n\n\ndef worker_cmd(py, worker, gen, inten, inp, out, device="cuda", batch_size=128, sparse_k=100, prec_margin=2.0):\n    return [str(py), str(worker), "--gen", str(gen), "--inten", str(inten), "--inp", str(inp), "--out", str(out),\n            "--device", device, "--batch-size", str(batch_size), "--sparse-k", str(sparse_k),\n            "--prec-margin", str(prec_margin)]\n\n\ndef read_response(out, n):\n    """-> spectra [(mz, it)], ok, err, timing"""\n    z = np.load(out)\n    # read each array ONCE: NpzFile re-reads (and re-allocates) an array on every z[key] access, and a\n    # slice keeps its whole copy alive -- per-row access held n full copies (the Phase-1 Colab OOM)\n    off, mz, it = z["off"], z["mz"], z["it"]\n    assert len(off) == n + 1, f"{out}: {len(off) - 1} rows, expected {n}"\n    spectra = [(mz[off[i]:off[i + 1]], it[off[i]:off[i + 1]]) for i in range(n)]\n    return spectra, z["ok"], z["err"], json.loads(str(z["timing"]))\n\n\ndef run_worker(req, work, tag, py, worker, gen, inten, env, **kw):\n    """-> spectra, ok, err, timing, wall_s"""\n    inp, out = Path(work) / f"req_{tag}.tsv", Path(work) / f"resp_{tag}.npz"\n    write_request(req, inp)\n    t = time.time()\n    r = subprocess.run(worker_cmd(py, worker, gen, inten, inp, out, **kw), text=True, capture_output=True,\n                       env=env, timeout=12 * 3600)\n    if r.returncode:\n        print(r.stdout[-2000:], r.stderr[-4000:])\n        raise RuntimeError(f"worker failed on {tag}")\n    spectra, ok, err, timing = read_response(out, len(req))\n    return spectra, ok, err, timing, time.time() - t\n\n\ndef err_counts(err):\n    return pd.Series([str(e).split(":")[0] for e in err if e]).value_counts().to_dict()\n\n\n# --- spectra -----------------------------------------------------------------------------------------\ndef prep(mz, it, prec):\n    """phase1 cleaning after the precursor + 2 Da cut. Needs the phase1 kernels in globals."""\n    mz = np.asarray(mz, np.float32)\n    it = np.asarray(it, np.float32)\n    keep = mz <= prec + 2.0\n    return _clean(mz[keep], it[keep], FLOOR, TOPK, POWER, ENTW)  # noqa: F821 (phase1 kernel cell)\n\n\ndef esim(a, b):\n    if len(a[0]) == 0 or len(b[0]) == 0:\n        return 0.0\n    return float(entropy_sim(a[0], a[1], b[0], b[1], TOL))  # noqa: F821 (phase1 kernel cell)\n\n\ndef sum_merge(specs, decimals=4):\n    """Merge per-CE predicted spectra the way a merged acquisition sums them.\n\n    Each input is (mz, it) with base peak 1. Peaks with the same m/z (rounded to `decimals`; the same\n    fragment formula gives the same m/z at every CE) are summed, then the result is renormalised to\n    base 1 and sorted by m/z. Empty inputs are ignored.\n    """\n    parts = [(np.asarray(m, np.float64), np.asarray(i, np.float64)) for m, i in specs if len(m)]\n    if not parts:\n        return np.empty(0, np.float32), np.empty(0, np.float32)\n    mz = np.concatenate([m for m, _ in parts])\n    it = np.concatenate([i / i.max() for _, i in parts])\n    key, inv = np.unique(np.round(mz, decimals), return_inverse=True)\n    s = np.bincount(inv, weights=it)\n    m = np.bincount(inv, weights=mz) / np.bincount(inv)\n    o = np.argsort(m)\n    return m[o].astype(np.float32), (s[o] / s.max()).astype(np.float32)\n\n\ndef rank_of_truth(sims):\n    """sims[0] = truth, sims[1:] = decoys. Ties count half; a tied truth is not top-1."""\n    sims = np.asarray(sims, dtype=np.float64)\n    t, dec = sims[0], sims[1:]\n    gt, eq = int((dec > t).sum()), int((dec == t).sum())\n    return 1 + gt + eq / 2, (gt == 0 and eq == 0)\n\n\ndef pb_sf(ps, k):\n    """P(X >= k), X = sum of independent Bernoulli(p_i) -- exact Poisson-binomial by DP."""\n    d = np.zeros(len(ps) + 1)\n    d[0] = 1.0\n    for p in ps:\n        d[1:] = d[1:] * (1 - p) + d[:-1] * p\n        d[0] *= (1 - p)\n    return float(d[k:].sum())\n\n\n# --- collision-energy grid ----------------------------------------------------------------------------\ndef ce_tuple(v):\n    """collision_energy_ev cell (list / array / None) -> tuple of floats rounded to 0.1 eV."""\n    if v is None:\n        return ()\n    try:\n        return tuple(round(float(x), 1) for x in v)\n    except TypeError:\n        return ()\n\n\ndef snap_ce(ce, grid, tol):\n    """Nearest grid CE within tol (eV), else None. Ties go to the lower CE."""\n    if not grid or not np.isfinite(ce):\n        return None\n    g = min(sorted(grid), key=lambda x: abs(x - ce))\n    return float(g) if abs(g - ce) <= tol + 1e-9 else None\n\n\ndef spectrum_conditions(adduct, ces, grid, tol, covered):\n    """One spectrum -> set of (adduct, ce, on_grid) conditions that need a simulated spectrum.\n\n    Uncovered adducts need none (MetFrag-lite). A merged multi-CE spectrum needs each constituent CE,\n    which is why merges add no grid conditions. CE <= 0 constituents (a few libraries list 0) are skipped.\n    """\n    if adduct not in covered:\n        return set()\n    out = set()\n    for ce in ces:\n        if not np.isfinite(ce) or ce <= 0:\n            continue\n        g = snap_ce(ce, grid, tol)\n        out.add((adduct, g if g is not None else round(float(ce), 1), g is not None))\n    return out\n\n\ndef choose_tolerance(median_by_delta, min_sim=0.90):\n    """Largest delta (eV) whose median sim(pred@c, pred@c+delta) is >= min_sim; 0.0 if none."""\n    ok = [d for d, m in median_by_delta.items() if np.isfinite(m) and m >= min_sim]\n    return float(max(ok)) if ok else 0.0\n\n\ndef base_ce_set(ce_tuples, min_cover=0.95):\n    """Distinct single CEs covering >= min_cover of single-CE spectra, plus every merged constituent."""\n    singles = pd.Series([t[0] for t in ce_tuples if len(t) == 1 and t[0] > 0])\n    keep = set()\n    if len(singles):\n        vc = singles.value_counts()\n        cum = vc.cumsum() / vc.sum()\n        keep = set(vc.index[: int((cum < min_cover).sum()) + 1])\n    for t in ce_tuples:\n        if len(t) > 1:\n            keep |= {c for c in t if c > 0}\n    return sorted(float(c) for c in keep)\n\n\n# --- candidate pool and windows ------------------------------------------------------------------------\ndef build_pool(coco_dir, train_path):\n    """COCONUT + train structures, deduplicated on ik14, sorted by exact neutral mass (as the gate).\n\n    Columns: k14, smiles, mass, has_real (the structure has >= 1 training spectrum).\n    """\n    coco_dir = Path(coco_dir)\n    cm = pickle.load(open(coco_dir / "coco_meta.pkl", "rb"))\n    cmass = np.load(coco_dir / "coco_mass.npy")\n    tr = pq.read_table(train_path, columns=["inchikey14", "normalized_smiles", "molecular_formula"]).to_pandas()\n    tr = tr.dropna(subset=["inchikey14", "normalized_smiles"]).drop_duplicates("inchikey14")\n    fm = {f: formula_mass(f) for f in set(tr.molecular_formula.astype(object).tolist())}\n    tr_mass = np.array([fm.get(f, np.nan) for f in tr.molecular_formula.astype(object)], dtype=np.float64)\n    pool = pd.DataFrame({\n        "k14": np.concatenate([[str(k)[:14] for k in cm["keys"]], tr.inchikey14.astype(str).to_numpy()]),\n        "smiles": np.concatenate([np.asarray(cm["smiles"], dtype=object),\n                                  tr.normalized_smiles.astype(object).to_numpy()]),\n        "mass": np.concatenate([cmass.astype(np.float64), tr_mass]),\n    })\n    n_raw = len(pool)\n    pool = (pool[np.isfinite(pool.mass) & pool.smiles.notna()].drop_duplicates("k14")\n            .sort_values("mass", kind="stable").reset_index(drop=True))\n    pool["has_real"] = pool.k14.isin(set(tr.inchikey14.astype(str)))\n    print(f"pool {n_raw:,} rows -> {len(pool):,} unique ik14 with a finite mass "\n          f"(COCONUT {len(cmass):,}, train {len(tr):,})")\n    return pool\n\n\ndef window(pm, target, ppm):\n    """(lo, hi) index range of the mass-sorted pool within +-ppm of target (phase1 CandidatePool.window)."""\n    lo = int(np.searchsorted(pm, target * (1 - ppm / 1e6), "left"))\n    hi = int(np.searchsorted(pm, target * (1 + ppm / 1e6), "right"))\n    return lo, hi\n\n\ndef scored_range(pm, target, win=PPM_WIN, fallback=PPM_FALLBACK, expand=PPM_EXPAND, topn=TOPN):\n    """Pool range whose candidates phase1\'s stages A + B actually score (stage C only mass-sorts).\n\n    Stage A: the +-win window, or +-fallback if that is empty. Stage B: while fewer than topn candidates\n    (pool keys are unique), every wider tier\'s new candidates are scored. The windows are nested, so the\n    scored set is one contiguous range.\n    """\n    lo, hi = window(pm, target, win)\n    if hi == lo:\n        lo, hi = window(pm, target, fallback)\n    if hi - lo < topn:\n        for ppm in expand[1:]:\n            a, b = window(pm, target, ppm)\n            lo, hi = min(lo, a), max(hi, b)\n            if hi - lo >= topn:\n                break\n    return lo, hi\n\n\ndef query_target(precs, adducts):\n    """Molecule -> centred neutral target mass (phase1: median over spectra, divided by 1 + PPM_CENTER)."""\n    nms = np.array([neutral_mass1(p, a) for p, a in zip(precs, adducts)], dtype=np.float64)\n    nms = nms[np.isfinite(nms)]\n    return float(np.median(nms)) / (1.0 + PPM_CENTER / 1e6) if len(nms) else np.nan\n\n\n# --- projection and decision ---------------------------------------------------------------------------\ndef on_the_fly(conds, precomputed_adducts):\n    """Conditions left for Kaggle: off-grid ones, and on-grid ones whose adduct was not precomputed."""\n    return [c for c in conds if not (c[2] and c[0] in precomputed_adducts)]\n\n\ndef bootstrap_total(per_mol, n_mol=400, n_boot=2000, seed=0):\n    """Total over n_mol molecules resampled from per_mol -> dict(mean, p50, p95)."""\n    x = np.asarray(per_mol, dtype=np.float64)\n    if len(x) == 0:\n        return dict(mean=0.0, p50=0.0, p95=0.0)\n    rng = np.random.default_rng(seed)\n    tot = x[rng.integers(0, len(x), size=(n_boot, n_mol))].sum(1)\n    return dict(mean=float(x.mean() * n_mol), p50=float(np.percentile(tot, 50)), p95=float(np.percentile(tot, 95)))\n\n\ndef kaggle_hours(load, rate, n_mol, t_base_s, fixed_s, simlib_gb, load_s_per_gb, worker_calls, startup_s):\n    """Projected Kaggle wall time (h): fixed setup + per-molecule base cost + simlib load + on-the-fly sims."""\n    sim_s = (load / rate if load > 0 else 0.0) + (worker_calls * startup_s if load > 0 else 0.0)\n    return (fixed_s + n_mol * t_base_s + simlib_gb * load_s_per_gb + sim_s) / 3600.0\n\n\ndef decide(options, budget_h):\n    """Cheapest option (colab_h, then name) whose kaggle_h <= budget_h; None if none fits (-> STOP)."""\n    fit = [o for o in options if o["kaggle_h"] <= budget_h]\n    return min(fit, key=lambda o: (o["colab_h"], o["name"])) if fit else None\n\n\n# --- submission (Phase 4, option A: every covered condition simulated on the fly) ---------------------\nMERGE_RULES = ("max", "sum", "mid")\n\n\ndef spectrum_ces(ces, rule, default):\n    """One query spectrum\'s CE list -> (CEs to simulate, defaulted).\n\n    CE <= 0 constituents are dropped (a few lists start at 0). `max` / `sum` simulate every constituent;\n    `mid` simulates the mean of the list once. No usable CE -> [default], defaulted = True.\n    """\n    assert rule in MERGE_RULES, rule\n    v = [round(float(c), 1) for c in ces if np.isfinite(c) and c > 0]\n    if not v:\n        return [float(default)], True\n    if rule == "mid":\n        return [round(float(np.mean(v)), 1)], False\n    return sorted(set(v)), False\n\n\ndef sim_priority(lv, fr, ppm):\n    """Per-candidate simulation priority (0 = first): lv desc, fr desc, |ppm| asc -- the watchdog\'s cap order."""\n    o = np.lexsort((np.asarray(ppm, np.float64), -np.asarray(fr, np.float64), -np.asarray(lv, np.float64)))\n    r = np.empty(len(o), np.int64)\n    r[o] = np.arange(len(o))\n    return r\n\n\nREQ_COLS = ["key", "smiles", "adduct", "precursor_mz", "ce_ev", "prio", "ce", "instrument"]\n\n\ndef build_requests(parts, instrument="QTOF", ce_mode="ev"):\n    """Molecule parts -> one worker request per distinct (key, adduct, ce_ev), sorted by priority.\n\n    Each part is a dict with arrays key / smiles / mass / prio (one entry per candidate to simulate) and\n    conds, a list of (adduct, ce_ev). The precursor is the candidate\'s theoretical m/z (exact mass +\n    adduct delta, as the G2b gate), so a request depends on the candidate alone and is shared by every\n    spectrum and molecule that needs it; a shared request keeps its best (lowest) priority.\n    """\n    frames = []\n    for p in parts:\n        if not len(p["key"]) or not p["conds"]:\n            continue\n        mass = np.asarray(p["mass"], np.float64)\n        for adduct, ce in p["conds"]:\n            frames.append(pd.DataFrame({"key": np.asarray(p["key"], dtype=object),\n                                        "smiles": np.asarray(p["smiles"], dtype=object), "adduct": adduct,\n                                        "precursor_mz": precursor_of(mass, adduct), "ce_ev": float(ce),\n                                        "prio": np.asarray(p["prio"], np.int64)}))\n    if not frames:\n        return pd.DataFrame({c: [] for c in REQ_COLS})\n    req = (pd.concat(frames, ignore_index=True)\n           .sort_values(["prio", "key", "adduct", "ce_ev"], kind="stable")\n           .drop_duplicates(["key", "adduct", "ce_ev"], keep="first").reset_index(drop=True))\n    req["ce"] = to_model_ce(req.ce_ev.to_numpy(), req.precursor_mz.to_numpy(), ce_mode)\n    req["instrument"] = instrument\n    return req[REQ_COLS]\n\n\ndef rows_that_fit(remaining_s, rate, startup_s):\n    """How many worker rows one more call can take before the deadline (0 if none)."""\n    return max(0, int((remaining_s - startup_s) * rate))\n\n\ndef spectrum_sv(q, sims, rule):\n    """Simulated-library score of ONE prepped query spectrum against one candidate.\n\n    sims: the candidate\'s raw predictions (mz, it, prec) at each of the spectrum\'s CEs, None where it was\n    not simulated (failed, capped). max / mid: best esim over the CEs; sum: esim against the renormalised\n    sum of the available CEs (a merged acquisition). Nothing available -> 0.\n    """\n    have = [s for s in sims if s is not None and len(s[0])]\n    if not have or len(q[0]) == 0:\n        return 0.0\n    if rule == "sum":\n        mz, it = sum_merge([(m, i) for m, i, _ in have])\n        return esim(q, prep(mz, it, have[0][2]))\n    return max(esim(q, prep(m, i, p)) for m, i, p in have)\n\n\ndef rank_two_tier(lv, sv, fr, ppm, tau):\n    """Order (indices) of one candidate set.\n\n    Tier 1 = real-library match lv >= tau, by lv desc (then sv, fr, |ppm|). Tier 2 = the rest, by sv desc,\n    fr desc, |ppm| asc. A missing sv / fr (NaN) counts 0; a missing ppm sorts last.\n    """\n    lv, sv, fr = (np.nan_to_num(np.asarray(x, np.float64), nan=0.0) for x in (lv, sv, fr))\n    ppm = np.nan_to_num(np.asarray(ppm, np.float64), nan=np.inf)\n    t1, t2 = np.flatnonzero(lv >= tau), np.flatnonzero(lv < tau)\n    o1 = t1[np.lexsort((ppm[t1], -fr[t1], -sv[t1], -lv[t1]))]\n    o2 = t2[np.lexsort((ppm[t2], -fr[t2], -sv[t2]))]\n    return np.concatenate([o1, o2]).astype(np.int64)\n'
NOVEL_COHORT_SRC = '"""specs_novel P1 helpers: honest benchmark cohorts and the decoder training split.\n\nThe definitions are pre-registered in specs_novel/gate_report.md -> G1 (design: phase1_plan.md §3-§4).\nnumpy / pandas plus simsim_common; RDKit and selfies are imported lazily (structure pass only).\nEmbedded verbatim into casmi_novel_p1_colab.ipynb by build/mk_novel_p0.py and imported directly by\nbuild/test_novel_p1.py.\n"""\nimport re\n\nimport numpy as np\nimport pandas as pd\n\nimport simsim_common as ssc\n\nNP_LIBS = ("riken", "mona", "msdial", "massbank", "gnps", "enveda-np-examples", "masaryk")\nTIMSTOF_LIB = "enveda-np-examples"\nDREAMS_FT_LIB, DREAMS_FT_ADDUCT = "mona", "[M+H]+"   # the DreaMS head\'s fine-tuning data (leakage flag)\nMASS_LO, MASS_HI = 157.0, 1159.0\nPPM_FLOOR, ROUND_TOL_DA = 10.0, 0.0075\nGMAX_CHOICES = (200, 100, 50, 25)\nHELDOUT_BUDGET = 0.07\nDEC_VAL_SHARE = 0.03\nN_C3, N_C2 = 600, 400\nMAX_SPEC = 16\nSEED = 20260929\nSPLITS = ("train", "dec_val", "heldout_c3", "heldout_c2", "heldout_ts")\n\n\n# --------------------------------------------------------------------------------------------- #\n# Per structure (RDKit, lazily)\n# --------------------------------------------------------------------------------------------- #\n\ndef scaffold_key(smiles, ik14):\n    """Murcko scaffold canonical SMILES of the stereo/isotope-stripped structure.\n    Empty scaffold -> \'acyclic:<ik14>\' (a group of one); unparsable -> \'unparsable:<ik14>\'."""\n    from rdkit import Chem, RDLogger\n    from rdkit.Chem.Scaffolds import MurckoScaffold\n    RDLogger.DisableLog("rdApp.*")\n    m = Chem.MolFromSmiles(smiles) if isinstance(smiles, str) and smiles else None\n    if m is None:\n        return f"unparsable:{ik14}"\n    Chem.RemoveStereochemistry(m)\n    for a in m.GetAtoms():\n        a.SetIsotope(0)\n    try:\n        s = Chem.MolToSmiles(MurckoScaffold.GetScaffoldForMol(m))\n    except Exception:\n        return f"unparsable:{ik14}"\n    return s if s else f"acyclic:{ik14}"\n\n\n_FTOK = re.compile(r"([A-Z][a-z]?)(\\d*)")\n\n\ndef formula_counts(f):\n    """\'C6H12O6\' -> {\'C\': 6, \'H\': 12, \'O\': 6}; None if the string is not a plain element formula."""\n    if not isinstance(f, str) or not f:\n        return None\n    out, pos = {}, 0\n    for m in _FTOK.finditer(f):\n        if m.start() != pos:\n            return None\n        pos = m.end()\n        out[m.group(1)] = out.get(m.group(1), 0) + (int(m.group(2)) if m.group(2) else 1)\n    return out if pos == len(f) else None\n\n\ndef structure_record(args):\n    """(ik14, normalized_smiles, molecular_formula, max_len) -> dict for one corpus structure.\n\n    prep via novel_decoder.prep_structure; `in_scope` = prep ok and SELFIES length <= max_len;\n    `formula_mismatch` = the RDKit formula of the prepped structure differs from the label formula\n    (None when prep did not produce a formula)."""\n    import selfies as sf\n    import novel_decoder as nd\n    ik, smi, label_formula, max_len = args\n    status, can, selfies, counts = nd.prep_structure(smi)\n    ntok = sf.len_selfies(selfies) if selfies else None\n    mismatch = None\n    if counts is not None:\n        lab = formula_counts(label_formula)\n        mismatch = lab is None or {k: v for k, v in lab.items() if v} != {k: v for k, v in counts.items() if v}\n    return dict(inchikey14=ik, status=status, can=can, selfies=selfies, ntok=ntok,\n                in_scope=bool(status == "ok" and ntok is not None and ntok <= max_len),\n                formula_mismatch=mismatch, scaffold_key=scaffold_key(smi, ik))\n\n\n# --------------------------------------------------------------------------------------------- #\n# Per row: label quality\n# --------------------------------------------------------------------------------------------- #\n\ndef label_ok(precursor_mz, adduct, formula):\n    """Vectorised G1 label-quality filter -> (ok bool array, neutral mass, truth mass).\n\n    ok = |neutral - truth| <= max(PPM_FLOOR ppm of truth, ROUND_TOL_DA) and neutral in [MASS_LO, MASS_HI].\n    Adducts outside the ten test adducts and unparsable formulas are not ok."""\n    mz = np.asarray(precursor_mz, np.float64)\n    delta = pd.Series(np.asarray(adduct, object)).map(ssc.ADDUCT_DELTA).to_numpy(np.float64)\n    f = pd.Series(np.asarray(formula, object))\n    uf = f.dropna().unique()\n    fm = f.map(dict(zip(uf, (ssc.formula_mass(x) for x in uf)))).to_numpy(np.float64)\n    neutral = mz - delta\n    tol = np.maximum(PPM_FLOOR * 1e-6 * fm, ROUND_TOL_DA)\n    with np.errstate(invalid="ignore"):\n        ok = (np.abs(neutral - fm) <= tol) & (neutral >= MASS_LO) & (neutral <= MASS_HI)\n    return ok & np.isfinite(neutral) & np.isfinite(fm), neutral, fm\n\n\n# --------------------------------------------------------------------------------------------- #\n# Per molecule: eligibility, group sizes, G_MAX\n# --------------------------------------------------------------------------------------------- #\n\ndef eligible_molecules(rows, msg_ik14):\n    """rows: DataFrame with inchikey14, ingest_lib, ok. -> one row per eligible ik14:\n    inchikey14, src_lib, timstof, n_ok (quality-OK rows in src_lib).\n\n    Eligible: ik14 not in msg_ik14 and >= 1 quality-OK row in an NP library. src_lib is the timsTOF\n    library when the molecule has quality-OK rows there, else its NP library with the most\n    quality-OK rows (ties -> the name that sorts first)."""\n    msg = set(np.asarray(msg_ik14).astype(str))\n    r = rows.loc[rows.ok.astype(bool) & rows.ingest_lib.isin(NP_LIBS), ["inchikey14", "ingest_lib"]]\n    r = r[~r.inchikey14.astype(str).isin(msg)]\n    cnt = r.groupby(["inchikey14", "ingest_lib"]).size().rename("n_ok").reset_index()\n    cnt["is_ts"] = cnt.ingest_lib == TIMSTOF_LIB\n    cnt = cnt.sort_values(["inchikey14", "is_ts", "n_ok", "ingest_lib"], ascending=[True, False, False, True])\n    out = cnt.drop_duplicates("inchikey14").rename(columns={"ingest_lib": "src_lib", "is_ts": "timstof"})\n    return out[["inchikey14", "src_lib", "timstof", "n_ok"]].reset_index(drop=True)\n\n\ndef group_sizes(scaf):\n    """scaf: Series ik14 -> scaffold_key (one entry per corpus ik14). -> Series ik14 -> group size."""\n    return scaf.map(scaf.value_counts()).astype(np.int64)\n\n\ndef choose_gmax(elig_group_size, n_universe, n_c3=N_C3, n_c2=N_C2, choices=GMAX_CHOICES,\n                budget=HELDOUT_BUDGET, dec_val=DEC_VAL_SHARE):\n    """Largest G in choices with n_c3 * m(G) + n_c2 + dec_val * U <= budget * U, where m(G) is the mean\n    group size of eligible molecules with group size <= G. -> (G, met, {G: m(G)}); if none meets the\n    budget, the smallest G with met = False."""\n    gs = np.asarray(elig_group_size, np.float64)\n    means = {int(g): (float(gs[gs <= g].mean()) if (gs <= g).any() else 0.0) for g in choices}\n    cap = budget * n_universe - n_c2 - dec_val * n_universe\n    for g in sorted(choices, reverse=True):\n        if n_c3 * means[g] <= cap:\n            return int(g), True, means\n    return int(min(choices)), False, means\n\n\n# --------------------------------------------------------------------------------------------- #\n# Cohorts\n# --------------------------------------------------------------------------------------------- #\n\ndef _draw(keys, n, rng):\n    a = np.array(sorted(keys), dtype=object)\n    if len(a) <= n:\n        return list(a)\n    return list(a[np.sort(rng.choice(len(a), size=n, replace=False))])\n\n\ndef sample_cohorts(elig, gmax, seed=SEED, n_c3=N_C3, n_c2=N_C2):\n    """elig: eligible_molecules() plus group_size and scaffold_key. -> (pseudo_c3 ik14s, c2_displace ik14s).\n\n    pseudo_c3: every eligible timsTOF molecule with group_size <= gmax (a uniform n_c3 of them if\n    more), then a uniform draw from the other eligible molecules with group_size <= gmax to n_c3.\n    c2_displace: a uniform n_c2 of the eligible molecules in no pseudo_c3 scaffold group (no cap)."""\n    rng = np.random.default_rng(seed)\n    cap = elig[elig.group_size <= gmax]\n    ts = set(cap.inchikey14[cap.timstof.astype(bool)])\n    c3 = _draw(ts, n_c3, rng)\n    rest = set(cap.inchikey14) - set(c3)\n    c3 += _draw(rest, n_c3 - len(c3), rng)\n    c3_scaf = set(elig.set_index("inchikey14").scaffold_key.loc[c3])\n    c2_pool = set(elig.inchikey14[~elig.scaffold_key.isin(c3_scaf)]) - set(c3)\n    c2 = _draw(c2_pool, n_c2, rng)\n    return sorted(c3), sorted(c2)\n\n\ndef sample_ts_dec(rows, msg_ik14, scaf, c3, c2, gmax):\n    """A-G1.1: the decoder-only timsTOF cohort. -> (sorted ts_dec ik14s, funnel dict).\n\n    Candidates: ik14 in msg_ik14 (ICEBERG has seen them) with >= 1 quality-OK timsTOF row. Excluded:\n    members of a pseudo_c3 scaffold group, c2_displace structures, members of a group holding a\n    c2_displace structure, and group size > gmax. Every remaining candidate is taken (no draw)."""\n    msg = set(np.asarray(msg_ik14).astype(str))\n    r = rows.loc[rows.ok.astype(bool) & (rows.ingest_lib == TIMSTOF_LIB), "inchikey14"].astype(str)\n    cand = set(r) & msg\n    size = scaf.value_counts()\n    c3_scaf, c2_scaf = set(scaf.loc[list(c3)]), set(scaf.loc[list(c2)])\n    after = {k for k in cand if scaf[k] not in c3_scaf and scaf[k] not in c2_scaf and k not in set(c2)}\n    ts = sorted(k for k in after if size[scaf[k]] <= gmax)\n    return ts, dict(candidates=len(cand), after_cohort_exclusions=len(after), within_gmax=len(ts))\n\n\ndef draw_n(spec_counts, rng, cap=MAX_SPEC):\n    """Spectra per molecule, drawn from the test\'s per-molecule counts, capped at `cap`."""\n    return int(min(cap, rng.choice(np.asarray(spec_counts))))\n\n\ndef pick_spectra(g, n, rng):\n    """g: candidate rows of one molecule (src_lib, quality-OK) with columns row, adduct, ce (NaN =\n    unknown). -> min(n, len(g)) row ids: CE-known rows first -- distinct (adduct, CE rounded to 0.1)\n    pairs, then the other CE-known rows -- and CE-unknown rows only to fill."""\n    g = g.sample(frac=1.0, random_state=int(rng.integers(1 << 30)))\n    ce = g.ce.to_numpy(np.float64)\n    known = g[np.isfinite(ce)]\n    unknown = g[~np.isfinite(ce)]\n    n = int(min(n, len(g)))\n    pick, seen = [], set()\n    for r in known.itertuples():\n        t = (r.adduct, round(float(r.ce), 1))\n        if t not in seen:\n            seen.add(t); pick.append(int(r.row))\n        if len(pick) == n:\n            return pick\n    chosen = set(pick)\n    for r in list(known.itertuples()) + list(unknown.itertuples()):\n        if len(pick) == n:\n            break\n        if int(r.row) not in chosen:\n            chosen.add(int(r.row)); pick.append(int(r.row))\n    return pick\n\n\ndef dreams_ft_risk(src_lib, adducts):\n    """True when the molecule\'s spectra come from the DreaMS head\'s fine-tuning data (MoNA [M+H]+)."""\n    return bool(src_lib == DREAMS_FT_LIB and DREAMS_FT_ADDUCT in set(adducts))\n\n\n# --------------------------------------------------------------------------------------------- #\n# Decoder training split\n# --------------------------------------------------------------------------------------------- #\n\ndef assign_splits(scaf, c3, c2, gmax, seed=SEED, dec_val=DEC_VAL_SHARE, ts=()):\n    """scaf: Series ik14 -> scaffold_key over every corpus ik14. -> Series ik14 -> split label.\n\n    heldout_c3: every member of a pseudo_c3 scaffold group; heldout_ts: every member of a ts_dec\n    scaffold group (A-G1.1); heldout_c2: the c2_displace structures; dec_val: whole random scaffold\n    groups with size <= gmax that hold no cohort structure, added in random order while the total\n    stays <= dec_val * U; train: the rest."""\n    rng = np.random.default_rng(seed + 1)\n    size = scaf.value_counts()\n    c3_scaf = set(scaf.loc[list(c3)])\n    ts_scaf = set(scaf.loc[list(ts)])\n    assert not ts_scaf & c3_scaf, "a ts_dec group is also a pseudo_c3 group"\n    split = pd.Series("train", index=scaf.index, dtype=object)\n    split[scaf.isin(c3_scaf)] = "heldout_c3"\n    split[scaf.isin(ts_scaf)] = "heldout_ts"\n    c2 = list(c2)\n    assert split.loc[c2].eq("train").all(), "a c2_displace structure is in a pseudo_c3 or ts_dec group"\n    split.loc[c2] = "heldout_c2"\n    c2_scaf = set(scaf.loc[c2])\n    busy = c3_scaf | ts_scaf | c2_scaf\n    cand = np.array(sorted(k for k, s in size.items() if s <= gmax and k not in busy),\n                    dtype=object)\n    rng.shuffle(cand)\n    target, total, take = dec_val * len(scaf), 0, []\n    for k in cand:\n        s = int(size[k])\n        if total + s <= target:\n            take.append(k); total += s\n        if total >= target - 0.5:\n            break\n    split[scaf.isin(set(take))] = "dec_val"\n    return split\n\n\ndef check_invariants(split, scaf, c3, c2, cohort_rows, row_ok, row_lib, src_lib_of, msg_ik14, ts=()):\n    """G1.4 (G1.4\' with ts_dec, A-G1.1) -> {invariant: bool}.\n\n    cohort_rows: {ik14: [corpus rows]} over every cohort; row_ok / row_lib: per corpus row;\n    src_lib_of: {ik14: src_lib}."""\n    msg = set(np.asarray(msg_ik14).astype(str))\n    c3, c2, ts = set(c3), set(c2), set(ts)\n    c3_members = set(scaf.index[scaf.isin(set(scaf.loc[list(c3)]))])\n    ts_members = set(scaf.index[scaf.isin(set(scaf.loc[list(ts)]))])\n    ok = {\n        "split_labels_valid": bool(split.isin(SPLITS).all()),\n        "split_covers_every_ik14_once": bool(split.index.is_unique and set(split.index) == set(scaf.index)),\n        "c3_groups_not_in_train_or_dec_val": bool(split.loc[list(c3_members)].eq("heldout_c3").all()),\n        "c2_not_in_train_or_dec_val": bool(split.loc[list(c2)].eq("heldout_c2").all()),\n        "cohort_rows_quality_ok_from_src_lib": all(\n            len(rs) > 0 and all(bool(row_ok[r]) and row_lib[r] == src_lib_of[ik] for r in rs)\n            for ik, rs in cohort_rows.items()),\n        "no_cohort_ik14_in_msg": not ((c3 | c2) & msg),\n        "cohorts_disjoint": not (c3 & c2),\n        "ts_groups_not_in_train_or_dec_val": bool(split.loc[list(ts_members)].eq("heldout_ts").all()),\n        "ts_disjoint_from_c3_c2": not (ts & (c3 | c2)),\n        "ts_all_in_msg": ts <= msg,\n        "ts_src_lib_timstof": all(src_lib_of[k] == TIMSTOF_LIB for k in ts),\n    }\n    return ok\n'

SRC_SHA1 = {}
for fname, text in [("novel_decoder.py", NOVEL_DECODER_SRC), ("simsim_common.py", SIMSIM_COMMON_SRC),
                    ("novel_cohort.py", NOVEL_COHORT_SRC)]:
    (SRC / fname).write_text(text, encoding="utf-8")
    SRC_SHA1[fname] = sha1_text(text)
sys.path.insert(0, str(SRC))
import novel_decoder as nd
import simsim_common as ssc
import novel_cohort as nc
nd.check_constraints()
assert nc.SEED == 20260929 and (nc.N_C3, nc.N_C2) == (600, 400) and nc.GMAX_CHOICES == (200, 100, 50, 25)
GATE_RESULT["src_sha1"] = SRC_SHA1
print(SRC_SHA1)
```

## Eligibility, `G_MAX` and the funnels

```python
ELIG = nc.eligible_molecules(META[["inchikey14", "ingest_lib", "ok"]], MSG)
ELIG = ELIG.merge(S[["inchikey14", "group_size", "scaffold_key", "in_scope"]], on="inchikey14", how="left")
assert ELIG.group_size.notna().all()
GMAX, GMAX_MET, M_G = nc.choose_gmax(ELIG.group_size.to_numpy(), len(S))
msg_set = set(MSG)

def funnel(rows):
    """molecules -> quality-OK in an NP library -> not in msg_ik14 -> group_size <= G_MAX."""
    k0 = set(rows.inchikey14)
    r = rows[rows.ok & rows.ingest_lib.isin(nc.NP_LIBS)]
    k1 = set(r.inchikey14)
    k2 = k1 - msg_set
    gs = S.set_index("inchikey14").group_size
    k3 = {k for k in k2 if gs[k] <= GMAX}
    return dict(molecules=len(k0), quality_ok_np=len(k1), not_in_msg=len(k2), within_gmax=len(k3))

FUNNEL = funnel(META)
FUNNEL_TS = funnel(META[META.ingest_lib == nc.TIMSTOF_LIB])
excl = float((ELIG.group_size > GMAX).mean())
GATE_RESULT["eligibility"] = dict(eligible=len(ELIG), eligible_timstof=int(ELIG.timstof.sum()),
                                  gmax=GMAX, gmax_met_budget=GMAX_MET, m_of_g=M_G,
                                  eligible_excluded_by_gmax=excl, funnel=FUNNEL, funnel_timstof=FUNNEL_TS,
                                  src_lib=ELIG.src_lib.value_counts().to_dict())
save_gate()
print(f"eligible {len(ELIG):,} (timsTOF {ELIG.timstof.sum()}); G_MAX {GMAX} (budget met: {GMAX_MET}); "
      f"m(G) {M_G}; excluded by G_MAX {100 * excl:.1f} %")
print(f"funnel {FUNNEL}\nfunnel timsTOF {FUNNEL_TS}")
```

## Cohorts and spectra per molecule

n per molecule is drawn from the test's spectra-per-molecule counts (capped at 16), rows come from
`src_lib` only, quality-OK, CE-known first, distinct (adduct, CE) pairs first. `ts_dec` (A-G1.1)
is drawn after `pseudo_c3` and `c2_displace`, so those two are the same as in run 1.

```python
C3, C2 = nc.sample_cohorts(ELIG, GMAX)
TS, TS_FUNNEL = nc.sample_ts_dec(META[["inchikey14", "ingest_lib", "ok"]], MSG, SCAF, C3, C2, GMAX)
SPEC_N = pd.read_parquet(TEST, columns=["molecule_id"]).molecule_id.value_counts().to_numpy()
E = ELIG.set_index("inchikey14")
SI = S.set_index("inchikey14")
SRC_OF = {**E.src_lib.to_dict(), **{k: nc.TIMSTOF_LIB for k in TS}}
CAND = META[META.ok & META.inchikey14.isin(set(C3) | set(C2) | set(TS))]
CAND = {k: g for k, g in CAND.groupby("inchikey14")}
LABEL_FORMULA = META.drop_duplicates("inchikey14").set_index("inchikey14").molecular_formula
rng = np.random.default_rng(nc.SEED + 2)

def cohort_frame(keys, prefix, pool_ban):
    recs = []
    for i, ik in enumerate(keys):
        src = SRC_OF[ik]
        g = CAND[ik]
        g = g[g.ingest_lib == src]
        rows = nc.pick_spectra(g[["row", "adduct", "ce"]], nc.draw_n(SPEC_N, rng), rng)
        sub = META.loc[rows]
        recs.append(dict(
            molecule_id=f"{prefix}{i:05d}", inchikey14=ik, smiles=SI.loc[ik, "normalized_smiles"],
            can=SI.loc[ik, "can"], selfies=SI.loc[ik, "selfies"], formula=LABEL_FORMULA[ik],
            scaffold=SI.loc[ik, "scaffold_key"], group_size=int(SI.loc[ik, "group_size"]), src_lib=src,
            timstof=src == nc.TIMSTOF_LIB, in_scope=bool(SI.loc[ik, "in_scope"]),
            rows=[int(r) for r in rows], train_rows=[int(x) for x in sub.train_row],
            adducts=list(sub.adduct), ces=[float(x) for x in sub.ce],
            lib_ban_keys=[ik], pool_ban_keys=[ik] if pool_ban else [],
            dreams_ft_risk=nc.dreams_ft_risk(src, sub.adduct)))
    return pd.DataFrame(recs)

T_C3 = cohort_frame(C3, "C3_", True)
T_C2 = cohort_frame(C2, "C2_", False)
T_TS = cohort_frame(TS, "TS_", True)               # decoder-only: G3 / G4.0, never ICEBERG metrics

def cohort_summary(T):
    ces = np.concatenate([np.asarray(c, np.float64) for c in T.ces]) if len(T) else np.zeros(0)
    return dict(n=len(T), timstof=int(T.timstof.sum()), in_scope_share=float(T.in_scope.mean()),
                dreams_ft_risk=int(T.dreams_ft_risk.sum()), by_src_lib=T.src_lib.value_counts().to_dict(),
                by_adduct=pd.Series([a for x in T.adducts for a in x]).value_counts().to_dict(),
                spectra=int(sum(len(r) for r in T.rows)), ce_known_share=float(np.isfinite(ces).mean()),
                spectra_per_mol=pd.Series([len(r) for r in T.rows]).describe().round(2).to_dict(),
                group_size_median=float(T.group_size.median()))

GATE_RESULT["cohorts"] = {"pseudo_c3": cohort_summary(T_C3), "c2_displace": cohort_summary(T_C2),
                          "ts_dec": {**cohort_summary(T_TS), "funnel": TS_FUNNEL}}
save_gate()
print(json.dumps(GATE_RESULT["cohorts"], indent=1))
```
