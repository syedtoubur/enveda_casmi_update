```python
'''all molecules -> base lists (v3.6 spectra construction and ordering)'''
te = pd.read_parquet(os.path.join(COMP, 'test.parquet'))
te['nm'] = chem.neutral_mass(te.precursor_mz.values.astype(np.float64), te.adduct.values)
mols = list(te.groupby('molecule_id', sort=False))
BASE, n_err = {}, 0
for gi, (mid, sub) in enumerate(mols):
    nms = sub.nm.values[np.isfinite(sub.nm.values)]
    smis, keys, lib_max, scs, forms = [], [], 0.0, [], []
    if len(nms):
        target = float(np.median(nms))
        spectra = []
        for r in sub.itertuples():
            ce = r.collision_energy_ev
            ce_arr = np.atleast_1d(np.asarray(ce, dtype=float)) if ce is not None else np.zeros(0)
            spectra.append(dict(mz=np.asarray(r.ms2_mzs, np.float64), it=np.asarray(r.ms2_normalized_intensities, np.float64),
                                mode=1 if r.ionization_mode == 'positive' else -1, adduct=r.adduct,
                                prec=float(r.precursor_mz), ce=float(ce_arr.mean()) if len(ce_arr) else np.nan,
                                ce_n=int(len(ce_arr)) if len(ce_arr) else 1))
        try:
            C, X, F, names, info = V.run(spectra, target)
            lib_max = float(info.get('lib_max', 0.0))
            if len(C.pid):
                score = rank_score(np.concatenate([X, F], 1), list(FEATURES) + list(names))
                order = np.argsort(-score, kind='stable')
                seen = set()
                for i in order:
                    k = C.key[i]
                    if k in seen: continue
                    seen.add(k); smis.append(C.smiles[i]); keys.append(k); scs.append(float(score[i])); forms.append(C.formula[i])
                    if len(smis) >= TOPN: break
        except Exception as e:
            print('ERROR', mid, repr(e)); n_err += 1
    BASE[mid] = (smis, keys, lib_max, scs, forms)
    if (gi + 1) % 25 == 0 or gi == len(mols) - 1:
        print(f'  {gi+1}/{len(mols)} molecules {time.time()-T0:.0f}s', flush=True)
print('errors', n_err)
```

```python
'''ICEBERG post-ranker isomer re-scoring (casmi26-iceberg). Optional: any failure keeps the ranker order.'''
import gc
ICE_SCORES, ice_stats = {}, dict(molecules=0, changed_top25=0, changed_top1=0)
try:
    ICE_PKG = os.path.dirname(find('casmi26-iceberg/**/ice_runner.py'))
    sys.path.append(ICE_PKG)
    import fuse as ice_fuse
    del V, E, bank; gc.collect(); torch.cuda.empty_cache()
    '''priority: molecules without a library-level match first (class 1 rarely changes), then the rest'''
    mids = sorted(BASE, key=lambda m: BASE[m][2])
    cands = {m: ice_fuse.ice_candidates(BASE[m][0], BASE[m][4], TOPN) for m in mids}
    if ENG:
        '''ADDED (ICE_UNION): also score our engine's candidates (same-formula groups'''
        from rdkit.Chem.rdMolDescriptors import CalcMolFormula as _CMF
        '''formed over the v4g top-N + engine top-40)'''
        from rdkit import Chem as _Ch
        def _f2(s):
            try:
                return _CMF(_Ch.MolFromSmiles(s))
            except Exception:
                return s
        n_add = 0
        for m in mids:
            e = ENG.get(str(m))
            if not e or not e.get('smiles'):
                continue
            bs, bk, bf = list(BASE[m][0][:TOPN]), set(BASE[m][1][:TOPN]), list(BASE[m][4][:TOPN])
            es = [s for s, k in zip(e['smiles'][:40], e['keys'][:40]) if k not in bk]
            ef = [_f2(s) for s in es]
            extra = ice_fuse.ice_candidates(bs + es, bf + ef, len(bs) + len(es))
            before = len(cands.get(m, []))
            cands[m] = list(dict.fromkeys(list(cands.get(m, [])) + extra))
            n_add += len(cands[m]) - before
        print('ICE_UNION: engine candidates added to ICE input:', n_add, flush=True)
    if ICE_PC:
        '''also the PubChem-only list of molecules the v3.6 gate may use (lib_max < LIB_TAU)'''
        from rdkit.Chem.rdMolDescriptors import CalcMolFormula
        from rdkit import Chem as _Chem
        def _form(s):
            try:
                return CalcMolFormula(_Chem.MolFromSmiles(s))
            except Exception:
                return s
        for m in mids:
            p = PC.get(m) if isinstance(PC, dict) else None
            if p and p.get('pc') and BASE[m][2] < LIB_TAU:
                p['pc_form'] = [_form(x) for x in p['pc']]
                extra = ice_fuse.ice_candidates(p['pc'], p['pc_form'], 25)
                cands[m] = list(dict.fromkeys(list(cands.get(m, [])) + extra))
    cands = {m: c for m, c in cands.items() if c}
    items = ice_fuse.build_ice_input(te, cands)
    print('ICE input molecules', len(items), 'candidates', sum(len(it['cands']) for it in items), f'{time.time()-T0:.0f}s', flush=True)
    ICE_SCORES = ice_fuse.run_ice(ICE_PKG, items, workdir='/kaggle/working/ice_work', device='cuda', budget_s=ICE_BUDGET,
                                  site='/kaggle/working/ice_site')
    meta_f = '/kaggle/working/ice_work/ice_out.json.meta.json'
    if os.path.exists(meta_f):
        print('ICE meta', open(meta_f).read()[:1500])
    '''ADDED: extra forward models on the same candidate sets, averaged with ICEBERG into one score per candidate'''
    FWD = {'ice': ICE_SCORES}
    if True:
        try:
            GL_PKG = os.path.dirname(find('casmi26-glacier/**/glacier_runner.py'))
            json.dump(items, open('/kaggle/working/gl_in.json', 'w'))
            _rg = subprocess.run([sys.executable, os.path.join(GL_PKG, 'glacier_runner.py'), GL_PKG, '/kaggle/working/gl_in.json',
                                  '/kaggle/working/gl_out.json', '--ice-pkg', ICE_PKG, '--device', 'cuda', '--budget', '3600',
                                  '--site', '/kaggle/working/gl_site', '--bs', '32'], capture_output=True, text=True, timeout=3600 + 900)
            print('GLACIER', _rg.stderr[-1500:], flush=True)
            FWD['gl'] = json.load(open('/kaggle/working/gl_out.json'))
        except Exception as _e:
            print('GLACIER FAILED', repr(_e))
    if False:
        try:
            FIO_DIR = '/kaggle/working/fio'; os.makedirs(FIO_DIR, exist_ok=True)
            open(os.path.join(FIO_DIR, 'fiora_score.py'), 'w').write('"""FIORA OS v1.0.0 scorer (MIT code; weights trained on MSnLib v7, CC BY 4.0).\n\nscore_candidates(queries, cands) -> {molecule_id: {smiles: mean entropy similarity or None}}\nOnly [M+H]+ and [M-H]- query spectra are scored (the only precursor types the OS model was trained on).\nCollision energy: FIORA was trained on Orbitrap NCE; timsTOF eV is converted with NCE = eV * 500 / precursor_mz.\n"""\nimport os, sys, glob, subprocess, time, warnings\nimport numpy as np\n\n_S = {}\nMODES = (\'[M+H]+\', \'[M-H]-\')\n\n\ndef setup(device=\'cuda\'):\n    if _S:\n        return\n    roots = sorted(glob.glob(\'/kaggle/input/**/casmi26-fiora\', recursive=True), key=len)\n    root = roots[0] if roots else \'/kaggle/input\'\n    whls = sorted(glob.glob(os.path.join(root, \'**\', \'*.whl\'), recursive=True))\n    r = subprocess.run([sys.executable, \'-m\', \'pip\', \'install\', \'-q\', \'--no-index\', \'--no-deps\'] + whls,\n                       capture_output=True, text=True)\n    print(\'fiora wheels\', [os.path.basename(w) for w in whls], r.returncode, r.stderr[-300:], flush=True)\n    fm = glob.glob(os.path.join(root, \'**\', \'fiora\', \'GNN\', \'FioraModel.py\'), recursive=True)[0]\n    src = os.path.dirname(os.path.dirname(os.path.dirname(fm)))\n    if src not in sys.path:\n        sys.path.insert(0, src)\n    warnings.filterwarnings(\'ignore\')\n    import torch, torch_geometric  # noqa: F401  (PyG must be imported before any torch_scatter shim)\n    from rdkit import RDLogger\n    RDLogger.DisableLog(\'rdApp.*\')\n    from fiora.GNN.FioraModel import FioraModel\n    from fiora.MS.SimulationFramework import SimulationFramework\n    from fiora.GNN.AtomFeatureEncoder import AtomFeatureEncoder\n    from fiora.GNN.BondFeatureEncoder import BondFeatureEncoder\n    from fiora.GNN.CovariateFeatureEncoder import CovariateFeatureEncoder\n    from fiora.MOL.Metabolite import Metabolite\n    state = glob.glob(os.path.join(root, \'**\', \'fiora_OS_v1.0.0_state.pt\'), recursive=True)[0]\n    model = FioraModel.load_from_state_dict(state.replace(\'_state.pt\', \'.pt\')).eval().to(device)\n    params = model.model_params\n    sets = params.get(\'setup_features_categorical_set\')\n    node = AtomFeatureEncoder(feature_list=[\'symbol\', \'num_hydrogen\', \'ring_type\'])\n    bond = BondFeatureEncoder(feature_list=[\'bond_type\', \'ring_type\'])\n    setup_enc = CovariateFeatureEncoder(feature_list=[\'collision_energy\', \'molecular_weight\', \'precursor_mode\',\n                                                      \'instrument\', \'element_composition\'], sets_overwrite=sets)\n    rt_enc = CovariateFeatureEncoder(feature_list=[\'molecular_weight\', \'precursor_mode\', \'instrument\'], sets_overwrite=sets)\n    setup_enc.normalize_features[\'collision_energy\'][\'max\'] = 100.0\n    setup_enc.normalize_features[\'molecular_weight\'][\'max\'] = 1000.0\n    rt_enc.normalize_features[\'molecular_weight\'][\'max\'] = 1000.0\n    _S.update(torch=torch, model=model, sim=SimulationFramework(None, dev=device), Metabolite=Metabolite,\n              node=node, bond=bond, setup_enc=setup_enc, rt_enc=rt_enc)\n    print(\'FIORA ready:\', params.get(\'version\'), \'on\', device, flush=True)\n\n\ndef _meta(adduct, nce):\n    return {\'name\': \'q\', \'collision_energy\': float(nce), \'instrument\': \'HCD\', \'precursor_mode\': adduct}\n\n\ndef make_metabolite(smi, adduct=\'[M+H]+\', nce=40.0):\n    try:\n        m = _S[\'Metabolite\'](smi)\n        m.create_molecular_structure_graph()\n        m.compute_graph_attributes(_S[\'node\'], _S[\'bond\'])\n        m.add_metadata(_meta(adduct, nce), _S[\'setup_enc\'], _S[\'rt_enc\'])\n        m.fragment_MOL(depth=1)\n        return m\n    except Exception:\n        return None\n\n\ndef predict(m, adduct, nce):\n    m.add_metadata(_meta(adduct, nce), _S[\'setup_enc\'], _S[\'rt_enc\'])\n    with _S[\'torch\'].no_grad():\n        out = _S[\'sim\'].simulate_and_score(m, _S[\'model\'], \'compiled_probsALL\', query_peaks=None)\n    p = out[\'sim_peaks\']\n    return np.asarray(p[\'mz\'], float), np.asarray(p[\'intensity\'], float)\n\n\n# ------------------------------------------------------------------------------------------ similarity\ndef _clean(mz, it, prec, floor=0.01, topk=150):\n    mz = np.asarray(mz, float); it = np.asarray(it, float)\n    k = (mz <= prec + 1.0) & np.isfinite(it) & (it > 0)\n    mz, it = mz[k], it[k]\n    if len(it) == 0:\n        return mz, it\n    k = it >= floor * it.max()\n    mz, it = mz[k], it[k]\n    if len(it) > topk:\n        o = np.argsort(-it)[:topk]; mz, it = mz[o], it[o]\n    o = np.argsort(mz)\n    return mz[o], it[o]\n\n\ndef _weight(p):\n    p = p / p.sum()\n    s = -np.sum(p * np.log(p))\n    if s < 3.0:\n        p = p ** (0.25 + 0.25 * s); p = p / p.sum()\n    return p\n\n\ndef entropy_sim(mz1, i1, mz2, i2, tol=0.01):\n    """Weighted spectral entropy similarity (Li et al. 2021), greedy peak matching within tol Da."""\n    if len(mz1) == 0 or len(mz2) == 0:\n        return 0.0\n    a = _weight(np.asarray(i1, float)); b = _weight(np.asarray(i2, float))\n    used = np.zeros(len(mz2), bool); merged = []\n    ua = np.zeros(len(mz1), bool)\n    for i in np.argsort(-a):\n        j0 = np.searchsorted(mz2, mz1[i] - tol); j1 = np.searchsorted(mz2, mz1[i] + tol, \'right\')\n        best, bj = -1.0, -1\n        for j in range(j0, j1):\n            if not used[j] and b[j] > best:\n                best, bj = b[j], j\n        if bj >= 0:\n            used[bj] = True; ua[i] = True; merged.append(a[i] + b[bj])\n    rest = np.concatenate([a[~ua], b[~used], np.asarray(merged)])\n    ab = rest / 2.0\n    ab = ab[ab > 0]\n    h = lambda p: -np.sum(p * np.log(p))\n    sa, sb, sab = h(a[a > 0]), h(b[b > 0]), h(ab)\n    return float(max(0.0, 1.0 - (2 * sab - sa - sb) / np.log(4)))\n\n\ndef nce_of(ce_ev, prec):\n    try:\n        v = np.atleast_1d(np.asarray(ce_ev, float))\n        ev = float(np.nanmean(v)) if len(v) else 30.0\n    except Exception:\n        ev = 30.0\n    if not np.isfinite(ev):\n        ev = 30.0\n    return float(np.clip(ev * 500.0 / max(float(prec), 50.0), 5.0, 100.0))\n\n\ndef score_candidates(queries, cands, max_cands=25, log_every=50, deadline=None, modes=MODES):\n    """queries: DataFrame with molecule_id, ms2_mzs, ms2_normalized_intensities, precursor_mz, adduct,\n    collision_energy_ev. cands: {mid: [smiles]} (put the molecules that matter most first; a deadline stops the loop).\n    modes: precursor types to score (subset of MODES). Returns {mid: {smiles: score or None}}."""\n    setup()\n    t0 = time.time(); out = {}; n_pred = 0\n    groups = {str(m): g for m, g in queries.groupby(\'molecule_id\')}\n    for gi, (mid, cs) in enumerate(cands.items()):\n        if deadline is not None and time.time() > deadline:\n            print(f\'fiora deadline reached after {gi} molecules\', flush=True)\n            break\n        g = groups.get(str(mid))\n        if g is None:\n            continue\n        specs = []\n        for r in g.itertuples():\n            if r.adduct not in modes:\n                continue\n            mz, it = _clean(r.ms2_mzs, r.ms2_normalized_intensities, r.precursor_mz)\n            if len(mz):\n                specs.append((r.adduct, round(nce_of(r.collision_energy_ev, r.precursor_mz), 1), mz, it))\n        if not specs:\n            continue\n        conds = sorted(set((a, n) for a, n, _, _ in specs))\n        res = {}\n        for smi in cs[:max_cands]:\n            m = make_metabolite(smi, *conds[0])\n            if m is None:\n                res[smi] = None; continue\n            pred = {}\n            try:\n                for a, n in conds:\n                    pred[(a, n)] = predict(m, a, n); n_pred += 1\n            except Exception:\n                res[smi] = None; continue\n            sims = []\n            for a, n, mz, it in specs:\n                pm, pi = pred[(a, n)]\n                o = np.argsort(pm)\n                sims.append(entropy_sim(mz, it, pm[o], pi[o]))\n            res[smi] = float(np.mean(sims)) if sims else None\n        out[mid] = res\n        if (gi + 1) % log_every == 0:\n            print(f\'  fiora {gi+1}/{len(cands)} molecules, {n_pred} predictions, {time.time()-t0:.0f}s\', flush=True)\n    print(f\'fiora done: {len(out)} molecules, {n_pred} predictions, {time.time()-t0:.0f}s\', flush=True)\n    return out\n')
            open(os.path.join(FIO_DIR, 'fiora_runner.py'), 'w').write('"""FIORA scoring in its own process (keeps torch_geometric out of the notebook process).\n\nusage: python fiora_runner.py <dir with fiora_score.py> <test.parquet> <cands.json> <out.json> <budget_s> [modes]\n  cands.json: {molecule_id: [smiles, ...]} in priority order;  modes: comma list, default "[M+H]+,[M-H]-"\nout.json: {molecule_id: {smiles: mean entropy similarity | null}}  (written even on failure)\n"""\nimport os, sys, json, time, traceback\n\nif __name__ == \'__main__\':\n    code_dir, test_path, cands_path, out_path, budget = sys.argv[1:6]\n    modes = tuple(sys.argv[6].split(\',\')) if len(sys.argv) > 6 else (\'[M+H]+\', \'[M-H]-\')\n    t0 = time.time()\n    json.dump({}, open(out_path, \'w\'))\n    try:\n        sys.path.insert(0, code_dir)\n        import pandas as pd\n        import fiora_score as FS\n        te = pd.read_parquet(test_path)\n        cands = json.load(open(cands_path))\n        res = FS.score_candidates(te, cands, max_cands=10 ** 6, deadline=t0 + float(budget), modes=modes)\n        json.dump(res, open(out_path, \'w\'))\n        print(\'fiora_runner ok\', len(res), f\'{time.time()-t0:.0f}s\', flush=True)\n    except Exception:\n        print(\'fiora_runner FAILED\', traceback.format_exc()[-3000:], flush=True)\n')
            json.dump(cands, open(os.path.join(FIO_DIR, 'cands.json'), 'w'))
            _rf = subprocess.run([sys.executable, os.path.join(FIO_DIR, 'fiora_runner.py'), FIO_DIR, os.path.join(COMP, 'test.parquet'),
                                  os.path.join(FIO_DIR, 'cands.json'), os.path.join(FIO_DIR, 'out.json'), '3600', '[M-H]-'],
                                 capture_output=True, text=True, timeout=3600 + 900)
            print('FIORA', _rf.stdout[-1500:], _rf.stderr[-800:], flush=True)
            FWD['fio'] = json.load(open(os.path.join(FIO_DIR, 'out.json')))
        except Exception as _e:
            print('FIORA FAILED', repr(_e))
    if len(FWD) > 1:
        _comb = {}
        for _mid in set().union(*[set(v) for v in FWD.values()]):
            _d = {}
            _pos_ok = any(_v is not None for _nm in ('ice', 'gl') for _v in ((FWD.get(_nm) or {}).get(_mid) or {}).values())
            for _nm, _w in (('ice', 1.0), ('gl', 1.0), ('fio', 1.0)):
                if _nm == 'fio' and True and _pos_ok:
                    continue
                for _s, _v in ((FWD.get(_nm) or {}).get(_mid) or {}).items():
                    if _v is None:
                        continue
                    _a = _d.setdefault(_s, [0.0, 0.0]); _a[0] += _w * _v; _a[1] += _w
            _comb[_mid] = {_s: (_a[0] / _a[1] if _a[1] > 0 else None) for _s, _a in _d.items()}
        ICE_SCORES = _comb
        print('forward models combined:', list(FWD), {k: len(v) for k, v in FWD.items()}, flush=True)
except Exception as e:
    print('ICE FAILED -> ranker order kept:', repr(e))
for m in list(BASE):
    smis, keys, lib_max, scs, forms = BASE[m]
    ice = ICE_SCORES.get(str(m), {}) if ICE_SCORES else {}
    if ice and any(v is not None for v in ice.values()):
        try:
            order = ice_fuse.rerank(smis, keys, scs, forms, ice, lam=ICE_LAM, top_n=TOPN)
            s2 = [smis[i] for i in order]; k2 = [keys[i] for i in order]
            ice_stats['molecules'] += 1
            ice_stats['changed_top25'] += int(s2[:25] != smis[:25]); ice_stats['changed_top1'] += int(bool(s2) and s2[0] != smis[0])
            smis, keys = s2, k2
        except Exception as e:
            print('rerank failed', m, repr(e))
    BASE[m] = (smis[:25], keys[:25], lib_max)
    p = PC.get(m) if (ICE_PC and isinstance(PC, dict)) else None
    if p and p.get('pc_form') and ice:
        try:
            '''re-order the PubChem list by z(f.z) + lam z(ice) inside same-formula groups;'''
            o = ice_fuse.rerank(p['pc'], p['pc_keys'], p['pc_fz'], p['pc_form'], ice, lam=ICE_LAM, top_n=25)
            fz0 = p['pc_fz'][0]
            '''the gate keeps using the ORIGINAL top f.z (rel), so gate decisions are unchanged'''
            p['pc'] = [p['pc'][i] for i in o]; p['pc_keys'] = [p['pc_keys'][i] for i in o]
            p['pc_fz'] = [fz0] + [p['pc_fz'][i] for i in o][1:]
            ice_stats['pc_changed'] = ice_stats.get('pc_changed', 0) + int(o[:1] != [0])
        except Exception as e:
            print('pc rerank failed', m, repr(e))
print('ICE rerank stats', ice_stats, f'{time.time()-T0:.0f}s')
```

```python
'''gated PubChem merge (v3.6, unchanged) -> submission.csv'''
def merge(base, base_keys, pc, pc_keys, slots, n=25):
    bk = set(base_keys)
    pcs = [s for s, k in zip(pc, pc_keys) if k not in bk]
    out, bi, pj = [], 0, 0
    for pos in range(1, n + 1):
        if pos in slots and pj < len(pcs):
            out.append(pcs[pj]); pj += 1
        elif bi < len(base):
            out.append(base[bi]); bi += 1
        elif pj < len(pcs):
            out.append(pcs[pj]); pj += 1
    return out
rows, stats = [], dict(untouched=0, gentle=0, aggressive=0, no_pc=0)
for mid in te.molecule_id.unique():
    smis, keys, lib_max = BASE.get(mid, ([], [], 0.0))
    p = PC.get(mid)
    if p is None or not p['pc']:
        stats['no_pc'] += 1; final = smis
    elif lib_max >= LIB_TAU:
        stats['untouched'] += 1; final = smis
    else:
        rel = p['pc_fz'][0] - p['best_pool_fz'] if np.isfinite(p['best_pool_fz']) else 1e9
        slots = SLOTS_AGG if rel > REL_TH else SLOTS_GENTLE
        stats['aggressive' if rel > REL_TH else 'gentle'] += 1
        final = merge(smis, keys, p['pc'], p['pc_keys'], slots)
    if not final: final = ['CCO']
    rows.append((mid, ';'.join(final[:25])))
print('merge stats', stats)
sub = pd.DataFrame(rows, columns=['molecule_id', 'smiles'])
samp = pd.read_csv(os.path.join(COMP, 'sample_submission.csv'))
sub = samp[['molecule_id']].merge(sub, on='molecule_id', how='left')
sub['smiles'] = sub['smiles'].fillna('CCO')
assert len(sub) == len(samp) and sub.molecule_id.is_unique and sub.smiles.notna().all()
assert sub.smiles.map(lambda s: len(s.split(';'))).max() <= 25
sub.to_csv('submission.csv', index=False)
json.dump(dict(secs=round(time.time() - T0), merge=stats, n_errors=n_err, families=MAN['families'],
               gpu=torch.cuda.get_device_name(0) if dev == 'cuda' else None), open('run_manifest.json', 'w'), indent=1)
print('done', sub.shape, f'{time.time()-T0:.0f}s')
sub.head()
```

```python

ALPHA, KRR = 0.6, 3.0
from rdkit.Chem.rdMolDescriptors import CalcMolFormula as _CMF2
from rdkit import Chem as _Ch2
def _form2(s):
    try:
        return _CMF2(_Ch2.MolFromSmiles(s))
    except Exception:
        return s
def fuse2(v_smis, e_smis, e_keys, n=40):
    sc, smi_of = {}, {}
    for r, s in enumerate(v_smis, 1):
        k = chem.score_key(s) or s
        sc[k] = sc.get(k, 0.0) + 1.0 / (KRR + r); smi_of.setdefault(k, s)
    for r, (s, k) in enumerate(zip(e_smis, e_keys), 1):
        if not k:
            continue
        sc[k] = sc.get(k, 0.0) + ALPHA / (KRR + r); smi_of.setdefault(k, s)
    order = sorted(sc, key=lambda k: -sc[k])[:n]
    return [smi_of[k] for k in order], order, [sc[k] for k in order]

if ENG:
    v4 = pd.read_csv('submission.csv')
    changed = n_ice = 0; out = []
    for mid, s in zip(v4.molecule_id, v4.smiles):
        vs = [x for x in s.split(';') if x and x != 'CCO']
        e = ENG.get(str(mid))
        if e and e['smiles']:
            fsm, fk, fsc = fuse2(vs, e['smiles'], e['keys'])
            ice = ICE_SCORES.get(str(mid), {}) if isinstance(ICE_SCORES, dict) else {}
            if ice:
                try:
                    o = ice_fuse.rerank(fsm, fk, fsc, [_form2(x) for x in fsm], ice, lam=ICE_LAM, top_n=len(fsm))
                    n_ice += int(o[:25] != list(range(min(25, len(o)))))
                    fsm = [fsm[i] for i in o]
                except Exception as ex:
                    print('fused ICE rerank failed', mid, repr(ex))
            changed += int(bool(vs) and fsm[:1] != vs[:1])
            vs = fsm[:25]
        out.append((mid, ';'.join(vs[:25]) if vs else 'CCO'))
    fs = pd.DataFrame(out, columns=['molecule_id', 'smiles'])
    assert len(fs) == len(v4) and fs.molecule_id.is_unique and fs.smiles.notna().all()
    assert fs.smiles.map(lambda s: len(s.split(';'))).max() <= 25
    fs.to_csv('submission.csv', index=False)
    print('fused (+ICE) submission written; top-1 changed in', changed, '| ICE reordered', n_ice, f'{time.time()-T0:.0f}s')
```
